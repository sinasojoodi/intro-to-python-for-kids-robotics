## 🎯 Puzzle 1: FizzBuzz

This is a famous programming challenge!

**Rules:**
- Count from 1 to 30
- If the number is divisible by 3, say "Fizz"
- If divisible by 5, say "Buzz"
- If divisible by both 3 and 5, say "FizzBuzz"
- Otherwise, say the number


```python
# Fibonacci Sequence!

print("=== FIBONACCI SEQUENCE ===\n")

def fibonacci(n):
    """Generate first n Fibonacci numbers"""
    fib_sequence = []
    a, b = 0, 1
    
    for i in range(n):
        fib_sequence.append(a)
        a, b = b, a + b  # Clever swap!
    
    return fib_sequence

# Generate first 15 Fibonacci numbers
fib_numbers = fibonacci(15)
print("First 15 Fibonacci numbers:")
print(fib_numbers)

# Pattern check
print("\n--- See the pattern? ---")
for i in range(len(fib_numbers)):
    print(f"F({i}) = {fib_numbers[i]}")

# Golden ratio
print("\n--- Golden Ratio ---")
print("As Fibonacci numbers get bigger,")
print("each number divided by the previous approaches 1.618...")
for i in range(5, 15):
    ratio = fib_numbers[i] / fib_numbers[i-1]
    print(f"{fib_numbers[i]} / {fib_numbers[i-1]} = {ratio:.3f}")
```

```python
# Multiplication Table Generator!

print("=== MULTIPLICATION TABLE ===\n")

size = 10

# Print header
print("    ", end="")
for i in range(1, size + 1):
    print(f"{i:4}", end="")
print()
print("    " + "-" * (size * 4))

# Print table
for i in range(1, size + 1):
    print(f"{i:2} |", end="")
    for j in range(1, size + 1):
        print(f"{i*j:4}", end="")
    print()

# Practice mode!
print("\n=== PRACTICE MODE ===")
import random

correct = 0
questions = 5

for q in range(questions):
    num1 = random.randint(1, 10)
    num2 = random.randint(1, 10)
    correct_answer = num1 * num2
    
    answer = int(input(f"\nWhat is {num1} × {num2}? "))
    
    if answer == correct_answer:
        print("✅ Correct!")
        correct += 1
    else:
        print(f"❌ No, it's {correct_answer}")

print(f"\nYou got {correct}/{questions} correct!")
```

## 🌟 Ultimate Challenge - Number Puzzle Solver!

Create a program that can:
1. Check if a number is perfect (sum of divisors equals the number)
2. Find all divisors of a number
3. Check if two numbers are "friendly" (their divisors sum to each other)
4. Find palindrome numbers (same forwards and backwards)

Be creative!


```python
# Number Puzzle Solver!

print("=== NUMBER PUZZLE SOLVER ===\n")

def get_divisors(n):
    """Find all divisors of a number"""
    divisors = []
    for i in range(1, n):
        if n % i == 0:
            divisors.append(i)
    return divisors

def is_perfect(n):
    """Check if number is perfect"""
    divisors = get_divisors(n)
    return sum(divisors) == n

def is_palindrome(n):
    """Check if number is palindrome"""
    str_n = str(n)
    return str_n == str_n[::-1]

# Test perfect numbers
print("Perfect numbers from 1 to 100:")
for num in range(1, 101):
    if is_perfect(num):
        print(f"{num} is perfect! Divisors: {get_divisors(num)}")

# Test palindromes
print("\nPalindrome numbers from 1 to 200:")
palindromes = [num for num in range(1, 201) if is_palindrome(num)]
print(palindromes)

# Interactive mode
print("\n=== INTERACTIVE MODE ===")
print("Enter a number to analyze: ")
number = int(input("Enter a number to analyze: "))

print(f"\nAnalyzing {number}...")
print(f"Divisors: {get_divisors(number)}")
print(f"Sum of divisors: {sum(get_divisors(number))}")
print(f"Perfect number? {is_perfect(number)}")
print(f"Palindrome? {is_palindrome(number)}")
print(f"Prime? {is_prime(number)}")  # Using function from earlier!
```

## 🔢 Puzzle 3: Prime Number Checker

A **prime number** is a number that can only be divided by 1 and itself.

Examples: 2, 3, 5, 7, 11, 13...

Let's write code to check if a number is prime!


```python
# Prime Number Checker!

def is_prime(number):
    if number < 2:
        return False
    
    # Check if any number from 2 to number-1 divides it
    for i in range(2, number):
        if number % i == 0:
            return False  # Found a divisor, not prime!
    
    return True  # No divisors found, it's prime!

# Test some numbers
print("=== PRIME NUMBER CHECKER ===\n")

test_numbers = [2, 3, 4, 5, 10, 11, 17, 20, 23]

for num in test_numbers:
    if is_prime(num):
        print(f"{num} is PRIME! ✅")
    else:
        print(f"{num} is not prime")

# Find all primes from 1 to 50
print("\n--- All primes from 1 to 50 ---")
primes = []
for num in range(1, 51):
    if is_prime(num):
        primes.append(num)

print(primes)
print(f"\nFound {len(primes)} prime numbers!")
```

## 🔢 Puzzle 4: Factorial Calculator

A **factorial** is when you multiply a number by all the numbers below it.

5! (5 factorial) = 5 × 4 × 3 × 2 × 1 = 120

Let's calculate factorials!


```python
# Factorial Calculator!

def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result

print("=== FACTORIAL CALCULATOR ===\n")

# Calculate some factorials
for num in range(1, 11):
    result = factorial(num)
    print(f"{num}! = {result}")

# Interactive calculator
print("\n--- Your Turn! ---")
print("Enter a number to find its factorial: ")
number = int(input("Enter a number to find its factorial: "))
result = factorial(number)
print(f"\n{number}! = {result}")

# Fun fact
print(f"\nThat's a {'small' if result < 1000 else 'BIG'} number!")
```

## 🌟 Challenge Corner: Math Tutor Program!

Create a complete math tutor that:
- Offers multiple operation types (add, subtract, multiply, divide)
- Multiple difficulty levels
- Tracks progress over multiple sessions
- Gives helpful hints
- Shows improvement stats!


```python
# MATH TUTOR PROGRAM!
import random

print("="*50)
print("📚 SMART MATH TUTOR 📚")
print("="*50)

# User stats
stats = {
    "correct": 0,
    "total": 0,
    "streak": 0
}

def get_problem(difficulty, operation):
    """Generate a math problem"""
    if difficulty == "easy":
        num1, num2 = random.randint(1, 10), random.randint(1, 10)
    elif difficulty == "medium":
        num1, num2 = random.randint(10, 50), random.randint(10, 50)
    else:  # hard
        num1, num2 = random.randint(50, 100), random.randint(50, 100)
    
    if operation == "+":
        answer = num1 + num2
    elif operation == "-":
        answer = num1 - num2
    elif operation == "*":
        answer = num1 * num2
    else:  # division
        answer = num1
        num1 = num1 * num2  # Ensure whole number answer
        operation = "÷"
    
    return num1, num2, operation, answer

def give_hint(correct_answer):
    """Give a helpful hint"""
    print(f"Hint: The answer is between {correct_answer - 5} and {correct_answer + 5}")

print("\nChoose difficulty:")
print("1. Easy (1-10)")
print("2. Medium (10-50)")
print("3. Hard (50-100)")
print("Enter 1, 2, or 3: ")
diff_choice = input("Enter 1, 2, or 3: ")
difficulty = ["easy", "medium", "hard"][int(diff_choice) - 1]

print("\nChoose operation:")
print("1. Addition (+)")
print("2. Subtraction (-)")
print("3. Multiplication (×)")
print("4. Division (÷)")
print("5. Mixed (random)")
print("Enter 1-5: ")
op_choice = input("Enter 1-5: ")
operations = ["+", "-", "*", "/"]
if op_choice != "5":
    operation = operations[int(op_choice) - 1]
else:
    operation = "mixed"

print("\nHow many questions? ")
num_questions = int(input("\nHow many questions? "))

print(f"\n{'='*50}")
print("Starting your math practice!")
print(f"{'='*50}\n")

for q in range(1, num_questions + 1):
    # Generate problem
    if operation == "mixed":
        current_op = random.choice(operations)
    else:
        current_op = operation
    
    num1, num2, op_symbol, correct_answer = get_problem(difficulty, current_op)
    
    print(f"Question {q}/{num_questions}:")
    print(f"{num1} {op_symbol} {num2} = ?")
    
    # Allow hint request
    print("Answer (or type 'hint'): ")
    user_input = input("Answer (or type 'hint'): ")
    
    if user_input.lower() == 'hint':
        give_hint(correct_answer)
        print("Answer: ")
        user_input = input("Answer: ")
    
    stats["total"] += 1
    
    try:
        user_answer = int(user_input)
        if user_answer == correct_answer:
            print("✅ Correct!\n")
            stats["correct"] += 1
            stats["streak"] += 1
        else:
            print(f"❌ Not quite. The answer was {correct_answer}\n")
            stats["streak"] = 0
    except:
        print("Invalid input!\n")
        stats["streak"] = 0

# Show final stats
print("="*50)
print("SESSION COMPLETE!")
print("="*50)
print(f"Correct: {stats['correct']}/{stats['total']}")
percentage = (stats['correct'] / stats['total']) * 100 if stats['total'] > 0 else 0
print(f"Score: {percentage:.1f}%")
print(f"Best streak: {stats['streak']}")

if percentage >= 90:
    print("\n🏆 AMAZING! You're a math genius!")
elif percentage >= 75:
    print("\n⭐ Great job! Keep it up!")
elif percentage >= 60:
    print("\n👍 Good work! Practice makes perfect!")
else:
    print("\n💪 Keep practicing! You'll get better!")
```
