## 🎯 Puzzle 1: FizzBuzzThis is a famous programming challenge!**Rules:**- Count from 1 to 30- If the number is divisible by 3, say "Fizz"- If divisible by 5, say "Buzz"- If divisible by both 3 and 5, say "FizzBuzz"- Otherwise, say the number

```python
# Fibonacci Sequence!print("=== FIBONACCI SEQUENCE ===\n")def fibonacci(n):    """Generate first n Fibonacci numbers"""    fib_sequence = []    a, b = 0, 1        for i in range(n):        fib_sequence.append(a)        a, b = b, a + b  # Clever swap!        return fib_sequence# Generate first 15 Fibonacci numbersfib_numbers = fibonacci(15)print("First 15 Fibonacci numbers:")print(fib_numbers)# Pattern checkprint("\n--- See the pattern? ---")for i in range(len(fib_numbers)):    print(f"F({i}) = {fib_numbers[i]}")# Golden ratioprint("\n--- Golden Ratio ---")print("As Fibonacci numbers get bigger,")print("each number divided by the previous approaches 1.618...")for i in range(5, 15):    ratio = fib_numbers[i] / fib_numbers[i-1]    print(f"{fib_numbers[i]} / {fib_numbers[i-1]} = {ratio:.3f}")
```

## 🌟 Ultimate Challenge - Number Puzzle Solver!Create a program that can:1. Check if a number is perfect (sum of divisors equals the number)2. Find all divisors of a number3. Check if two numbers are "friendly" (their divisors sum to each other)4. Find palindrome numbers (same forwards and backwards)Be creative!

```python
# Number Puzzle Solver!print("=== NUMBER PUZZLE SOLVER ===\n")def get_divisors(n):    """Find all divisors of a number"""    divisors = []    for i in range(1, n):        if n % i == 0:            divisors.append(i)    return divisorsdef is_perfect(n):    """Check if number is perfect"""    divisors = get_divisors(n)    return sum(divisors) == ndef is_palindrome(n):    """Check if number is palindrome"""    str_n = str(n)    return str_n == str_n[::-1]# Test perfect numbersprint("Perfect numbers from 1 to 100:")for num in range(1, 101):    if is_perfect(num):        print(f"{num} is perfect! Divisors: {get_divisors(num)}")# Test palindromesprint("\nPalindrome numbers from 1 to 200:")palindromes = [num for num in range(1, 201) if is_palindrome(num)]print(palindromes)# Interactive modeprint("\n=== INTERACTIVE MODE ===")number = int(input("Enter a number to analyze: "))print(f"\nAnalyzing {number}...")print(f"Divisors: {get_divisors(number)}")print(f"Sum of divisors: {sum(get_divisors(number))}")print(f"Perfect number? {is_perfect(number)}")print(f"Palindrome? {is_palindrome(number)}")print(f"Prime? {is_prime(number)}")  # Using function from earlier!
```

## 🔢 Puzzle 3: Prime Number CheckerA **prime number** is a number that can only be divided by 1 and itself.Examples: 2, 3, 5, 7, 11, 13...Let's write code to check if a number is prime!

```python
# Prime Number Checker!def is_prime(number):    if number < 2:        return False        # Check if any number from 2 to number-1 divides it    for i in range(2, number):        if number % i == 0:            return False  # Found a divisor, not prime!        return True  # No divisors found, it's prime!# Test some numbersprint("=== PRIME NUMBER CHECKER ===\n")test_numbers = [2, 3, 4, 5, 10, 11, 17, 20, 23]for num in test_numbers:    if is_prime(num):        print(f"{num} is PRIME! ✅")    else:        print(f"{num} is not prime")# Find all primes from 1 to 50print("\n--- All primes from 1 to 50 ---")primes = []for num in range(1, 51):    if is_prime(num):        primes.append(num)print(primes)print(f"\nFound {len(primes)} prime numbers!")
```

## 🔢 Puzzle 4: Factorial CalculatorA **factorial** is when you multiply a number by all the numbers below it.5! (5 factorial) = 5 × 4 × 3 × 2 × 1 = 120Let's calculate factorials!

```python
# Factorial Calculator!def factorial(n):    result = 1    for i in range(1, n + 1):        result = result * i    return resultprint("=== FACTORIAL CALCULATOR ===\n")# Calculate some factorialsfor num in range(1, 11):    result = factorial(num)    print(f"{num}! = {result}")# Interactive calculatorprint("\n--- Your Turn! ---")number = int(input("Enter a number to find its factorial: "))result = factorial(number)print(f"\n{number}! = {result}")# Fun factprint(f"\nThat's a {'small' if result < 1000 else 'BIG'} number!")
```
