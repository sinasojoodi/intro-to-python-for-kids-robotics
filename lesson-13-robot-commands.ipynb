## 🎨 Visual Grid Robot

Let's create a robot that moves on a visual grid!


```python
# Grid Robot with Obstacles!

class GridRobot:
    def __init__(self, grid_size=8):
        self.size = grid_size
        self.x = 0
        self.y = 0
        self.grid = [['.' for _ in range(grid_size)] for _ in range(grid_size)]
        self.obstacles = []
        self.treasures = []
        self.collected = []
        
    def add_obstacle(self, x, y):
        """Add an obstacle"""
        self.obstacles.append((x, y))
        self.grid[y][x] = '#'
    
    def add_treasure(self, x, y):
        """Add treasure"""
        self.treasures.append((x, y))
        self.grid[y][x] = '$'
    
    def show_grid(self):
        """Display the grid"""
        print("\n" + "=" * (self.size * 2 + 1))
        for y in range(self.size - 1, -1, -1):
            for x in range(self.size):
                if (x, y) == (self.x, self.y):
                    print("🤖", end="")
                elif self.grid[y][x] == '#':
                    print("🧱", end="")
                elif self.grid[y][x] == '$':
                    print("💎", end="")
                else:
                    print("  ", end="")
            print()
        print("=" * (self.size * 2 + 1))
    
    def move(self, direction):
        """Move the robot"""
        new_x, new_y = self.x, self.y
        
        if direction == "up":
            new_y = min(self.size - 1, self.y + 1)
        elif direction == "down":
            new_y = max(0, self.y - 1)
        elif direction == "left":
            new_x = max(0, self.x - 1)
        elif direction == "right":
            new_x = min(self.size - 1, self.x + 1)
        
        # Check for obstacles
        if (new_x, new_y) in self.obstacles:
            print("❌ Can't move there! Obstacle!")
            return False
        
        self.x, self.y = new_x, new_y
        
        # Check for treasure
        if (self.x, self.y) in self.treasures and (self.x, self.y) not in self.collected:
            print("💎 Treasure found!")
            self.collected.append((self.x, self.y))
            self.grid[self.y][self.x] = '.'
        
        return True

# Play the grid game!
print("=== GRID ROBOT GAME ===")

robot = GridRobot(8)

# Add obstacles
robot.add_obstacle(2, 2)
robot.add_obstacle(3, 2)
robot.add_obstacle(4, 2)

# Add treasures
robot.add_treasure(5, 5)
robot.add_treasure(7, 7)
robot.add_treasure(1, 6)

robot.show_grid()

# Interactive mode
print("\nCommands: up, down, left, right, quit")
print(f"Goal: Collect all {len(robot.treasures)} treasures!\n")

while len(robot.collected) < len(robot.treasures):
    print("Enter command: ")
    command = input("Enter command: ").lower()
    
    if command == "quit":
        break
    elif command in ["up", "down", "left", "right"]:
        robot.move(command)
        robot.show_grid()
        print(f"Treasures collected: {len(robot.collected)}/{len(robot.treasures)}")
    else:
        print("Invalid command!")

if len(robot.collected) == len(robot.treasures):
    print("\n🎉 You collected all treasures! Winner!")
```

```python
# Simple Robot Simulator!

class Robot:
    def __init__(self, name):
        self.name = name
        self.x = 0
        self.y = 0
        self.direction = "North"
        self.holding = None
        
    def move_forward(self):
        if self.direction == "North":
            self.y += 1
        elif self.direction == "South":
            self.y -= 1
        elif self.direction == "East":
            self.x += 1
        elif self.direction == "West":
            self.x -= 1
        print(f"{self.name} moves forward to ({self.x}, {self.y})")
    
    def turn_left(self):
        directions = ["North", "West", "South", "East"]
        current_index = directions.index(self.direction)
        self.direction = directions[(current_index + 1) % 4]
        print(f"{self.name} turns left, now facing {self.direction}")
    
    def turn_right(self):
        directions = ["North", "East", "South", "West"]
        current_index = directions.index(self.direction)
        self.direction = directions[(current_index + 1) % 4]
        print(f"{self.name} turns right, now facing {self.direction}")
    
    def pick_up(self, item):
        self.holding = item
        print(f"{self.name} picks up {item}")
    
    def put_down(self):
        if self.holding:
            print(f"{self.name} puts down {self.holding}")
            self.holding = None
        else:
            print(f"{self.name} is not holding anything!")
    
    def status(self):
        print(f"\n--- {self.name} Status ---")
        print(f"Position: ({self.x}, {self.y})")
        print(f"Facing: {self.direction}")
        print(f"Holding: {self.holding if self.holding else 'Nothing'}\n")

# Create a robot!
print("=== ROBOT SIMULATOR ===\n")

robo = Robot("Robo-Owen")
robo.status()

# Give it commands!
robo.move_forward()
robo.move_forward()
robo.turn_right()
robo.move_forward()
robo.pick_up("treasure")
robo.turn_left()
robo.move_forward()
robo.put_down()

robo.status()
```

## 🧭 Robot Maze Navigator

Let's make a robot that can navigate through a maze!


```python
# Maze Navigator Robot!

class MazeRobot:
    def __init__(self):
        self.x = 0
        self.y = 0
        self.path = [(0, 0)]
        
    def move(self, direction):
        if direction == "up":
            self.y += 1
        elif direction == "down":
            self.y -= 1
        elif direction == "left":
            self.x -= 1
        elif direction == "right":
            self.x += 1
        
        self.path.append((self.x, self.y))
        print(f"Moved {direction} to ({self.x}, {self.y})")
    
    def show_path(self):
        print("\n📍 Robot's Path:")
        for i, pos in enumerate(self.path):
            print(f"Step {i}: {pos}")
    
    def draw_path(self):
        """Draw a simple ASCII map of the path"""
        if not self.path:
            return
        
        # Find bounds
        min_x = min(p[0] for p in self.path)
        max_x = max(p[0] for p in self.path)
        min_y = min(p[1] for p in self.path)
        max_y = max(p[1] for p in self.path)
        
        print("\n🗺️ Map of Journey:")
        for y in range(max_y, min_y - 1, -1):
            for x in range(min_x, max_x + 1):
                if (x, y) == (0, 0):
                    print("S", end=" ")  # Start
                elif (x, y) == self.path[-1]:
                    print("E", end=" ")  # End
                elif (x, y) in self.path:
                    print("*", end=" ")  # Path
                else:
                    print(".", end=" ")  # Empty
            print()

# Navigate a maze!
print("=== MAZE NAVIGATOR ===\n")

robot = MazeRobot()

# Mission: Go around a square and return home
print("Mission: Navigate around a 3x3 square!\n")

robot.move("right")
robot.move("right")
robot.move("right")
robot.move("up")
robot.move("up")
robot.move("up")
robot.move("left")
robot.move("left")
robot.move("left")
robot.move("down")
robot.move("down")
robot.move("down")

robot.show_path()
robot.draw_path()
```

## 🎮 Robot Functions

Instead of giving the robot the same commands over and over, we can create FUNCTIONS!

This is like teaching the robot new tricks!


```python
# Teaching Robot New Tricks!

class SmartRobot:
    def __init__(self, name):
        self.name = name
        self.x = 0
        self.y = 0
        
    def move(self, direction, steps=1):
        for _ in range(steps):
            if direction == "up":
                self.y += 1
            elif direction == "down":
                self.y -= 1
            elif direction == "left":
                self.x -= 1
            elif direction == "right":
                self.x += 1
        print(f"{self.name} moved {steps} steps {direction} to ({self.x}, {self.y})")
    
    def make_square(self, size):
        """Robot draws a square!"""
        print(f"\n{self.name} is drawing a square of size {size}!")
        start_x, start_y = self.x, self.y
        
        self.move("right", size)
        self.move("up", size)
        self.move("left", size)
        self.move("down", size)
        
        print(f"Square complete! Back at ({self.x}, {self.y})")
    
    def go_home(self):
        """Return to (0, 0)"""
        print(f"\n{self.name} is going home...")
        while self.x != 0 or self.y != 0:
            if self.x > 0:
                self.move("left")
            elif self.x < 0:
                self.move("right")
            elif self.y > 0:
                self.move("down")
            elif self.y < 0:
                self.move("up")
        print("Home sweet home! 🏠")
    
    def dance(self):
        """Robot does a little dance!"""
        print(f"\n{self.name} is dancing! 💃")
        moves = ["right", "left", "left", "right", "up", "down"]
        for move in moves:
            self.move(move)
        print("Dance complete!")

# Test the smart robot!
print("=== SMART ROBOT ===\n")

robot = SmartRobot("SmartBot")

robot.make_square(3)
robot.dance()
robot.go_home()
```

## 📦 Delivery Robot Game

Let's create a game where a robot has to deliver packages!


```python
# Delivery Robot Game!
import random

class DeliveryRobot:
    def __init__(self):
        self.x = 0
        self.y = 0
        self.energy = 100
        self.packages_delivered = 0
        
    def move_to(self, target_x, target_y):
        """Move to a specific location"""
        steps = 0
        
        while self.x != target_x or self.y != target_y:
            if self.energy <= 0:
                print("❌ Out of energy! Game over!")
                return False
            
            if self.x < target_x:
                self.x += 1
                print(f"Moving right... ({self.x}, {self.y})")
            elif self.x > target_x:
                self.x -= 1
                print(f"Moving left... ({self.x}, {self.y})")
            elif self.y < target_y:
                self.y += 1
                print(f"Moving up... ({self.x}, {self.y})")
            elif self.y > target_y:
                self.y -= 1
                print(f"Moving down... ({self.x}, {self.y})")
            
            self.energy -= 5
            steps += 1
        
        print(f"✅ Arrived at ({self.x}, {self.y})!")
        print(f"Steps taken: {steps}, Energy left: {self.energy}")
        return True
    
    def deliver_package(self, location):
        """Deliver a package to a location"""
        print(f"\n📦 New delivery to {location}!")
        target_x, target_y = location
        
        if self.move_to(target_x, target_y):
            self.packages_delivered += 1
            print(f"📬 Package delivered! Total: {self.packages_delivered}")
            
            # Return home
            print("\n🏠 Returning home...")
            self.move_to(0, 0)
            return True
        return False
    
    def recharge(self):
        print("\n⚡ Recharging...")
        self.energy = 100
        print("Energy full!")

# Play the delivery game!
print("=== DELIVERY ROBOT GAME ===\n")

robot = DeliveryRobot()

# Generate random delivery locations
deliveries = [
    (3, 2),
    (5, 5),
    (2, 4),
]

for i, location in enumerate(deliveries, 1):
    print(f"\n🎯 Delivery {i} of {len(deliveries)}")
    if robot.energy < 30:
        robot.recharge()
    
    success = robot.deliver_package(location)
    if not success:
        print("Mission failed!")
        break
else:
    print(f"\n🎉 All deliveries complete!")
    print(f"Total packages delivered: {robot.packages_delivered}")
    print("You're an amazing robot programmer!")
```

## 🌟 Challenge Corner: Robot Soccer!

Build a complete robot soccer game with:
- Two robot players
- A ball that moves
- Goal detection
- Score keeping
- Turn-based gameplay!


```python
# ROBOT SOCCER GAME!
import random

print("="*50)
print("⚽ ROBOT SOCCER CHAMPIONSHIP ⚽")
print("="*50)

class SoccerRobot:
    def __init__(self, name, position):
        self.name = name
        self.x, self.y = position
        self.score = 0
    
    def move_towards(self, target_x, target_y):
        """Move one step towards target"""
        if self.x < target_x:
            self.x += 1
        elif self.x > target_x:
            self.x -= 1
        elif self.y < target_y:
            self.y += 1
        elif self.y > target_y:
            self.y -= 1
        print(f"  {self.name} moves to ({self.x}, {self.y})")
    
    def kick_ball(self, ball, direction):
        """Kick the ball"""
        print(f"  {self.name} kicks {direction}!")
        if direction == "left":
            ball.x -= 2
        elif direction == "right":
            ball.x += 2
        elif direction == "up":
            ball.y += 2
        elif direction == "down":
            ball.y -= 2

class Ball:
    def __init__(self):
        self.x, self.y = 5, 5  # Center of field
    
    def reset(self):
        self.x, self.y = 5, 5

def show_field(robot1, robot2, ball):
    """Display the soccer field"""
    print("\n" + "="*30)
    for y in range(10, -1, -1):
        for x in range(11):
            if (x, y) == (robot1.x, robot1.y):
                print("🔴", end="")
            elif (x, y) == (robot2.x, robot2.y):
                print("🔵", end="")
            elif (x, y) == (ball.x, ball.y):
                print("⚽", end="")
            elif x == 0 or x == 10:
                print("|", end=" ")
            else:
                print(".", end=" ")
        if y == 5:
            print(" 🥅")
        else:
            print()
    print("="*30)

# Initialize game
robot1 = SoccerRobot("Red Bot", (2, 5))
robot2 = SoccerRobot("Blue Bot", (8, 5))
ball = Ball()

print("\nRed Bot's goal is on the left (x=0)")
print("Blue Bot's goal is on the right (x=10)\n")

# Simple AI game
for turn in range(15):
    print(f"\n--- Turn {turn + 1} ---")
    show_field(robot1, robot2, ball)
    
    # Robot 1's turn (trying to score on right)
    if abs(robot1.x - ball.x) + abs(robot1.y - ball.y) <= 1:
        robot1.kick_ball(ball, "right")
    else:
        robot1.move_towards(ball.x, ball.y)
    
    # Check goal
    if ball.x >= 10:
        robot1.score += 1
        print("\n🎉 RED BOT SCORES!")
        ball.reset()
    elif ball.x <= 0:
        robot2.score += 1
        print("\n🎉 BLUE BOT SCORES!")
        ball.reset()
    
    # Robot 2's turn (trying to score on left)
    if abs(robot2.x - ball.x) + abs(robot2.y - ball.y) <= 1:
        robot2.kick_ball(ball, "left")
    else:
        robot2.move_towards(ball.x, ball.y)
    
    # Check goal again
    if ball.x >= 10:
        robot1.score += 1
        print("\n🎉 RED BOT SCORES!")
        ball.reset()
    elif ball.x <= 0:
        robot2.score += 1
        print("\n🎉 BLUE BOT SCORES!")
        ball.reset()

# Final score
show_field(robot1, robot2, ball)
print("\n" + "="*50)
print("FINAL SCORE")
print("="*50)
print(f"Red Bot: {robot1.score}")
print(f"Blue Bot: {robot2.score}")

if robot1.score > robot2.score:
    print("\n🏆 RED BOT WINS!")
elif robot2.score > robot1.score:
    print("\n🏆 BLUE BOT WINS!")
else:
    print("\n🤝 IT'S A TIE!")

print("\nGreat game! 🎮")
```
