## 🎨 Visual Grid RobotLet's create a robot that moves on a visual grid!

```python
# Grid Robot with Obstacles!class GridRobot:    def __init__(self, grid_size=8):        self.size = grid_size        self.x = 0        self.y = 0        self.grid = [['.' for _ in range(grid_size)] for _ in range(grid_size)]        self.obstacles = []        self.treasures = []        self.collected = []            def add_obstacle(self, x, y):        """Add an obstacle"""        self.obstacles.append((x, y))        self.grid[y][x] = '#'        def add_treasure(self, x, y):        """Add treasure"""        self.treasures.append((x, y))        self.grid[y][x] = '$'        def show_grid(self):        """Display the grid"""        print("\n" + "=" * (self.size * 2 + 1))        for y in range(self.size - 1, -1, -1):            for x in range(self.size):                if (x, y) == (self.x, self.y):                    print("🤖", end="")                elif self.grid[y][x] == '#':                    print("🧱", end="")                elif self.grid[y][x] == '$':                    print("💎", end="")                else:                    print("  ", end="")            print()        print("=" * (self.size * 2 + 1))        def move(self, direction):        """Move the robot"""        new_x, new_y = self.x, self.y                if direction == "up":            new_y = min(self.size - 1, self.y + 1)        elif direction == "down":            new_y = max(0, self.y - 1)        elif direction == "left":            new_x = max(0, self.x - 1)        elif direction == "right":            new_x = min(self.size - 1, self.x + 1)                # Check for obstacles        if (new_x, new_y) in self.obstacles:            print("❌ Can't move there! Obstacle!")            return False                self.x, self.y = new_x, new_y                # Check for treasure        if (self.x, self.y) in self.treasures and (self.x, self.y) not in self.collected:            print("💎 Treasure found!")            self.collected.append((self.x, self.y))            self.grid[self.y][self.x] = '.'                return True# Play the grid game!print("=== GRID ROBOT GAME ===")robot = GridRobot(8)# Add obstaclesrobot.add_obstacle(2, 2)robot.add_obstacle(3, 2)robot.add_obstacle(4, 2)# Add treasuresrobot.add_treasure(5, 5)robot.add_treasure(7, 7)robot.add_treasure(1, 6)robot.show_grid()# Interactive modeprint("\nCommands: up, down, left, right, quit")print(f"Goal: Collect all {len(robot.treasures)} treasures!\n")while len(robot.collected) < len(robot.treasures):    command = input("Enter command: ").lower()        if command == "quit":        break    elif command in ["up", "down", "left", "right"]:        robot.move(command)        robot.show_grid()        print(f"Treasures collected: {len(robot.collected)}/{len(robot.treasures)}")    else:        print("Invalid command!")if len(robot.collected) == len(robot.treasures):    print("\n🎉 You collected all treasures! Winner!")
```

## 🧭 Robot Maze NavigatorLet's make a robot that can navigate through a maze!

```python
# Maze Navigator Robot!class MazeRobot:    def __init__(self):        self.x = 0        self.y = 0        self.path = [(0, 0)]            def move(self, direction):        if direction == "up":            self.y += 1        elif direction == "down":            self.y -= 1        elif direction == "left":            self.x -= 1        elif direction == "right":            self.x += 1                self.path.append((self.x, self.y))        print(f"Moved {direction} to ({self.x}, {self.y})")        def show_path(self):        print("\n📍 Robot's Path:")        for i, pos in enumerate(self.path):            print(f"Step {i}: {pos}")        def draw_path(self):        """Draw a simple ASCII map of the path"""        if not self.path:            return                # Find bounds        min_x = min(p[0] for p in self.path)        max_x = max(p[0] for p in self.path)        min_y = min(p[1] for p in self.path)        max_y = max(p[1] for p in self.path)                print("\n🗺️ Map of Journey:")        for y in range(max_y, min_y - 1, -1):            for x in range(min_x, max_x + 1):                if (x, y) == (0, 0):                    print("S", end=" ")  # Start                elif (x, y) == self.path[-1]:                    print("E", end=" ")  # End                elif (x, y) in self.path:                    print("*", end=" ")  # Path                else:                    print(".", end=" ")  # Empty            print()# Navigate a maze!print("=== MAZE NAVIGATOR ===\n")robot = MazeRobot()# Mission: Go around a square and return homeprint("Mission: Navigate around a 3x3 square!\n")robot.move("right")robot.move("right")robot.move("right")robot.move("up")robot.move("up")robot.move("up")robot.move("left")robot.move("left")robot.move("left")robot.move("down")robot.move("down")robot.move("down")robot.show_path()robot.draw_path()
```

## 🎮 Robot FunctionsInstead of giving the robot the same commands over and over, we can create FUNCTIONS!This is like teaching the robot new tricks!

```python
# Teaching Robot New Tricks!class SmartRobot:    def __init__(self, name):        self.name = name        self.x = 0        self.y = 0            def move(self, direction, steps=1):        for _ in range(steps):            if direction == "up":                self.y += 1            elif direction == "down":                self.y -= 1            elif direction == "left":                self.x -= 1            elif direction == "right":                self.x += 1        print(f"{self.name} moved {steps} steps {direction} to ({self.x}, {self.y})")        def make_square(self, size):        """Robot draws a square!"""        print(f"\n{self.name} is drawing a square of size {size}!")        start_x, start_y = self.x, self.y                self.move("right", size)        self.move("up", size)        self.move("left", size)        self.move("down", size)                print(f"Square complete! Back at ({self.x}, {self.y})")        def go_home(self):        """Return to (0, 0)"""        print(f"\n{self.name} is going home...")        while self.x != 0 or self.y != 0:            if self.x > 0:                self.move("left")            elif self.x < 0:                self.move("right")            elif self.y > 0:                self.move("down")            elif self.y < 0:                self.move("up")        print("Home sweet home! 🏠")        def dance(self):        """Robot does a little dance!"""        print(f"\n{self.name} is dancing! 💃")        moves = ["right", "left", "left", "right", "up", "down"]        for move in moves:            self.move(move)        print("Dance complete!")# Test the smart robot!print("=== SMART ROBOT ===\n")robot = SmartRobot("SmartBot")robot.make_square(3)robot.dance()robot.go_home()
```
