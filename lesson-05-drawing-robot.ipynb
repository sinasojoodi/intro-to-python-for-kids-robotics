## ⚡ Turtle Speed SettingsYou can make your turtle go faster or slower!- `speed(1)` = slowest (like a real turtle!)- `speed(5)` = medium- `speed(10)` = fastest- `speed(0)` = SUPER FAST (no animation!)

## 🌟 Drawing Challenge!Can you draw a flower? Here's a hint:- Use circles for petals- Draw multiple circles in different positions- Use `goto(x, y)` to move to different spots- Remember to `penup()` before moving, `pendown()` before drawing!

```python
# Draw a flower!import turtleartist = turtle.Turtle()artist.speed(5)# Draw petals (circles) around a center point# Try drawing 5-6 circles in different spots!# Example: first petalartist.circle(40)# Move to new position for next petalartist.penup()artist.goto(50, 50)artist.pendown()# Draw more petals!# You can do it!turtle.done()
```

Today you learned:- ✅ How to use Python's turtle graphics- ✅ Commands like `forward()`, `right()`, `circle()`- ✅ How to draw shapes (squares, triangles, circles)- ✅ How to control the pen with `penup()` and `pendown()`**Next lesson:** Rainbow Art - adding colors and loops! 🌈You're an artist AND a programmer now!

```python
# Wake up the turtle!import turtle# Create a turtle named "artist"artist = turtle.Turtle()# Let's draw!artist.forward(100)  # Move forward 100 stepsartist.right(90)     # Turn right 90 degreesartist.forward(100)# Keep the window openturtle.done()
```

## 🎮 Turtle CommandsHere are the main commands to control your turtle:**Movement:**- `forward(distance)` - move forward- `backward(distance)` - move backward- `right(angle)` - turn right (in degrees)- `left(angle)` - turn left (in degrees)**Pen Control:**- `penup()` - lift the pen (move without drawing)- `pendown()` - put pen down (start drawing again)Think of it like a remote control for your robot!

## ⬛ Let's Draw a Square!A square has 4 equal sides and 4 right angles (90 degrees). Let's draw one!

```python
# Drawing a square!import turtleartist = turtle.Turtle()artist.speed(2)  # Slow so we can watch!# Draw 4 sidesartist.forward(100)artist.right(90)artist.forward(100)artist.right(90)artist.forward(100)artist.right(90)artist.forward(100)artist.right(90)print("Look! A square!")turtle.done()
```

## 🔺 Drawing a TriangleA triangle has 3 sides. When turning, we need to turn 120 degrees for each corner!

```python
# Drawing a triangle!import turtleartist = turtle.Turtle()artist.speed(2)# Draw 3 sidesartist.forward(100)artist.right(120)artist.forward(100)artist.right(120)artist.forward(100)artist.right(120)print("A triangle!")turtle.done()
```

## ⭕ Drawing a CircleCircles are easy! The turtle has a special `circle()` command:

```python
# Drawing circles!import turtleartist = turtle.Turtle()artist.speed(3)# Draw a circleartist.circle(50)  # The number is the radius (size)# Move and draw anotherartist.penup()artist.goto(100, 0)artist.pendown()artist.circle(30)print("Circles!")turtle.done()
```
