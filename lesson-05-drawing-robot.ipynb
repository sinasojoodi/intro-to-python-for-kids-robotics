## 🎮 Turtle Commands

Here are the main commands to control your turtle:

**Movement:**
- `forward(distance)` - move forward
- `backward(distance)` - move backward
- `right(angle)` - turn right (in degrees)
- `left(angle)` - turn left (in degrees)

**Pen Control:**
- `penup()` - lift the pen (move without drawing)
- `pendown()` - put pen down (start drawing again)

Think of it like a remote control for your robot!

## 📍 Positioning Practice: goto()

The `goto(x, y)` command moves the turtle to an exact spot!

- (0, 0) is the center of the screen
- Positive x goes right, negative x goes left
- Positive y goes up, negative y goes down

```python
# Drawing in specific locations!
import turtle
artist = turtle.Turtle()
artist.speed(5)

# Draw a square in the top-right
artist.penup()
artist.goto(100, 100)
artist.pendown()
for i in range(4):
    artist.forward(50)
    artist.right(90)

# Draw a circle in the top-left
artist.penup()
artist.goto(-100, 100)
artist.pendown()
artist.circle(30)

# Draw a triangle in the bottom-right
artist.penup()
artist.goto(100, -100)
artist.pendown()
for i in range(3):
    artist.forward(60)
    artist.right(120)

# YOU draw a shape in the bottom-left at (-100, -100)!
# Your code here!

turtle.done()
```

## 🏠 Mini-Project: Building a House!

Let's build a house step-by-step! A house has:
1. A square for the main part
2. A triangle for the roof
3. A rectangle for the door
4. Squares for windows


```python
# Building a House!
import turtle
artist = turtle.Turtle()
artist.speed(3)

# Step 1: Draw the main square (house body)
print("Drawing house body...")
for i in range(4):
    artist.forward(150)
    artist.right(90)

# Step 2: Draw the roof (triangle)
print("Drawing roof...")
artist.left(30)  # Adjust angle for roof
artist.forward(87)
artist.right(120)
artist.forward(174)
artist.right(120)
artist.forward(87)

# Step 3: Draw the door
print("Drawing door...")
artist.penup()
artist.goto(55, 0)
artist.right(90)
artist.pendown()
# Door is a tall rectangle
artist.forward(60)
artist.right(90)
artist.forward(40)
artist.right(90)
artist.forward(60)
artist.right(90)
artist.forward(40)

# Step 4: Draw windows
print("Drawing windows...")
# Left window
artist.penup()
artist.goto(20, 60)
artist.pendown()
for i in range(4):
    artist.forward(25)
    artist.right(90)

# Right window - YOU draw this!
# Put it at position (105, 60)
# Make it 25x25
# Your code here!

print("House complete!")
turtle.done()
```

## 🎯 Practice Time: Pattern Creation!

Let's create patterns by repeating shapes!


```python
# Drawing a FENCE pattern!
import turtle
artist = turtle.Turtle()
artist.speed(7)

print("Building a fence...")

# Draw 5 fence posts
for post in range(5):
    # Draw one post (tall rectangle)
    for side in range(2):
        artist.forward(80)  # tall
        artist.right(90)
        artist.forward(10)   # narrow
        artist.right(90)
    
    # Move to next post position
    artist.penup()
    artist.forward(30)
    artist.pendown()

print("Fence complete!")
turtle.done()
```

```python
# Drawing a BRICK WALL pattern!
import turtle
artist = turtle.Turtle()
artist.speed(8)

print("Building a brick wall...")

# Starting position
start_x = -200
start_y = 100

# Draw 3 rows of bricks
for row in range(3):
    # Offset every other row
    if row % 2 == 0:
        offset = 0
    else:
        offset = 30
    
    # Go to start of row
    artist.penup()
    artist.goto(start_x + offset, start_y - (row * 40))
    artist.pendown()
    
    # Draw 5 bricks in this row
    for brick in range(5):
        # Draw one brick (rectangle)
        for side in range(2):
            artist.forward(60)
            artist.right(90)
            artist.forward(30)
            artist.right(90)
        
        # Move to next brick
        artist.penup()
        artist.forward(60)
        artist.pendown()

print("Wall complete!")
turtle.done()
```

## 🌟 Challenge Corner: Complete Scene!

Ready for the ultimate challenge? Draw a complete scene with:
- A house
- A tree
- The sun
- Clouds

Let's do it step by step!


```python
# COMPLETE SCENE!
import turtle
artist = turtle.Turtle()
artist.speed(8)

print("Creating a beautiful scene...")

# === SUN (top right corner) ===
print("Drawing sun...")
artist.penup()
artist.goto(150, 150)
artist.pendown()
artist.circle(30)

# === CLOUDS (top) ===
print("Drawing clouds...")
artist.penup()
artist.goto(-150, 120)
artist.pendown()
# Cloud made of 3 circles
artist.circle(20)
artist.penup()
artist.forward(25)
artist.pendown()
artist.circle(20)
artist.penup()
artist.forward(25)
artist.pendown()
artist.circle(20)

# === HOUSE (center-left) ===
print("Drawing house...")
artist.penup()
artist.goto(-100, -50)
artist.pendown()
# House body
for i in range(4):
    artist.forward(100)
    artist.right(90)
# Roof
artist.left(45)
artist.forward(71)
artist.right(90)
artist.forward(71)

# Door
artist.penup()
artist.goto(-70, -50)
artist.right(135)
artist.pendown()
for i in range(2):
    artist.forward(40)
    artist.right(90)
    artist.forward(25)
    artist.right(90)

# Window
artist.penup()
artist.goto(-80, 0)
artist.pendown()
for i in range(4):
    artist.forward(20)
    artist.right(90)

# === TREE (right side) ===
print("Drawing tree...")
# Tree trunk
artist.penup()
artist.goto(80, -50)
artist.pendown()
artist.right(90)
for i in range(2):
    artist.forward(60)
    artist.right(90)
    artist.forward(20)
    artist.right(90)

# Tree leaves (circle on top)
artist.penup()
artist.goto(90, 10)
artist.pendown()
artist.circle(40)

# === GROUND (optional - a line) ===
print("Drawing ground...")
artist.penup()
artist.goto(-250, -50)
artist.pendown()
artist.forward(500)

# === YOUR TURN! ===
# Add more to the scene!
# Ideas:
# - More clouds
# - Flowers (circle for flower, stem is a line)
# - A path to the house
# - Stars
# - Birds (simple V shapes)
# Your code here!

print("Scene complete!")
turtle.done()
```
