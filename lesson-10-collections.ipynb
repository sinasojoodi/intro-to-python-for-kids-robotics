## 🎯 Accessing List ItemsEach item in a list has a position number called an **index**.**Important:** Python counts from 0, not 1!- First item: index 0- Second item: index 1- Third item: index 2Use brackets to get items: `my_list[0]`

````python
# Accessing items!fruits = ["apple", "banana", "orange", "grape", "mango"]print("First fruit:", fruits[0])print("Second fruit:", fruits[1])print("Last fruit:", fruits[4])# Negative indexes count from the end!print("Last fruit (easy way):", fruits[-1])print("Second to last:", fruits[-2])# Get the length (how many items)print("\nTotal fruits:", len(fruits))# Get a slice (range of items)print("First 3 fruits:", fruits[0:3])print("Last 2 fruits:", fruits[-2:])
````

## ✏️ Modifying ListsLists are special because you CAN change them!**Common operations:**- `list.append(item)` - add item to the end- `list.insert(index, item)` - add item at position- `list.remove(item)` - remove specific item- `list.pop()` - remove and return last item- `list[index] = new_value` - change an item

````python
# Modifying lists!toys = ["ball", "car", "puzzle"]print("Starting toys:", toys)# Add to the endtoys.append("robot")print("After append:", toys)# Add at specific positiontoys.insert(1, "doll")  # Add at index 1print("After insert:", toys)# Change an itemtoys[0] = "super ball"print("After changing:", toys)# Remove an itemtoys.remove("puzzle")print("After remove:", toys)# Remove last itemlast_toy = toys.pop()print(f"Removed: {last_toy}")print("Final toys:", toys)
````

## 🔁 Looping Through ListsRemember loops? They're PERFECT for going through lists!

````python
# Looping through lists!colors = ["red", "blue", "green", "yellow", "purple"]print("My favorite colors:")for color in colors:    print(f"  - {color}")print("\n--- With index numbers ---")for i in range(len(colors)):    print(f"{i + 1}. {colors[i]}")print("\n--- Using enumerate ---")for index, color in enumerate(colors, start=1):    print(f"Color #{index}: {color}")
````
