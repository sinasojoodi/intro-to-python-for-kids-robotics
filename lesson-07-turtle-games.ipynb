## ⌨️ Keyboard Control!

You can make your turtle move when you press keys on the keyboard! This is how games work!

We'll use:
- `onkey(function, "key")` - tells turtle what to do when a key is pressed
- `listen()` - tells turtle to listen for keyboard


```python
# Random walker!
import turtle
import random

artist = turtle.Turtle()
artist.speed(0)
screen = turtle.Screen()
screen.bgcolor("white")

colors = ["red", "blue", "green", "yellow", "purple", "orange", "pink"]

for i in range(100):
    # Pick random color
    artist.color(random.choice(colors))
    
    # Random size
    size = random.randint(10, 100)
    artist.circle(size)
    
    # Random direction
    artist.right(random.randint(0, 360))
    
    # Random distance
    artist.penup()
    artist.forward(random.randint(10, 50))
    artist.pendown()

turtle.done()
```

## ✏️ Your Turn - Draw by Clicking!

Let's make a program where you click the screen and the turtle draws there!


```python
# Click to draw!
import turtle

screen = turtle.Screen()
screen.bgcolor("lightyellow")
artist = turtle.Turtle()
artist.speed(0)
artist.shape("circle")

# Function that runs when you click
def draw_dot(x, y):
    artist.penup()
    artist.goto(x, y)
    artist.pendown()
    artist.dot(20, "red")

# Listen for clicks
screen.onclick(draw_dot)

print("Click anywhere to draw dots!")
turtle.done()
```

## 🏁 Turtle Race Game!

Let's create a simple racing game with two turtles!


```python
# Turtle race!
import turtle
import random

screen = turtle.Screen()
screen.bgcolor("lightgreen")
screen.title("Turtle Race!")

# Create racing turtles
turtle1 = turtle.Turtle()
turtle1.color("red")
turtle1.shape("turtle")
turtle1.penup()
turtle1.goto(-200, 50)

turtle2 = turtle.Turtle()
turtle2.color("blue")
turtle2.shape("turtle")
turtle2.penup()
turtle2.goto(-200, -50)

# Draw finish line
finish = turtle.Turtle()
finish.penup()
finish.goto(200, 100)
finish.pendown()
finish.right(90)
finish.forward(200)

# Race!
print("Racing...")
for i in range(100):
    turtle1.forward(random.randint(1, 10))
    turtle2.forward(random.randint(1, 10))

# Who won?
if turtle1.xcor() > turtle2.xcor():
    print("RED turtle wins!")
else:
    print("BLUE turtle wins!")

turtle.done()
```

## 🌟 Ultimate Challenge - Drawing App!

Create your own drawing app! Features to include:
- Arrow keys to move the turtle
- 'c' key to change color
- 'u' key to pick pen up
- 'd' key to put pen down
- 's' key to stamp
- Space bar to clear screen

Make it YOUR way!


```python
# Your drawing app!
import turtle
import random

screen = turtle.Screen()
screen.bgcolor("white")
screen.title("Owen's Drawing App")

artist = turtle.Turtle()
artist.speed(0)

colors = ["red", "blue", "green", "yellow", "purple", "orange", "pink", "black"]
color_index = 0

# Movement functions
def forward():
    artist.forward(20)

def backward():
    artist.backward(20)

def left():
    artist.left(45)

def right():
    artist.right(45)

def change_color():
    global color_index
    color_index = (color_index + 1) % len(colors)
    artist.color(colors[color_index])
    print("Color:", colors[color_index])

def pen_up():
    artist.penup()
    print("Pen up!")

def pen_down():
    artist.pendown()
    print("Pen down!")

def stamp_it():
    artist.stamp()

def clear_all():
    artist.clear()
    print("Screen cleared!")

# Connect keys
screen.onkey(forward, "Up")
screen.onkey(backward, "Down")
screen.onkey(left, "Left")
screen.onkey(right, "Right")
screen.onkey(change_color, "c")
screen.onkey(pen_up, "u")
screen.onkey(pen_down, "d")
screen.onkey(stamp_it, "s")
screen.onkey(clear_all, "space")

screen.listen()

print("=== DRAWING APP ===")
print("Arrow keys: Move")
print("C: Change color")
print("U/D: Pen up/down")
print("S: Stamp")
print("Space: Clear")

turtle.done()
```

## 🖱️ Advanced Click Handling!

You can do more than just draw dots! Let's make clicks do cool things!


```python
# Click to draw colored circles!
import turtle
import random

screen = turtle.Screen()
screen.bgcolor("white")
artist = turtle.Turtle()
artist.speed(0)
artist.hideturtle()

colors = ["red", "orange", "yellow", "green", "blue", "purple", "pink"]
click_count = 0

def draw_circle(x, y):
    global click_count
    click_count += 1
    
    artist.penup()
    artist.goto(x, y)
    artist.pendown()
    
    # Pick a color
    color = colors[click_count % len(colors)]
    artist.color(color)
    
    # Draw a filled circle
    artist.begin_fill()
    artist.circle(30)
    artist.end_fill()
    
    print(f"Circle #{click_count} at ({x}, {y})")

screen.onclick(draw_circle)
print("Click to draw colorful circles!")
turtle.done()
```

```python
# Click to draw shapes!
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)

shapes = ["square", "triangle", "circle", "star"]
current_shape = 0

def draw_at_click(x, y):
    global current_shape
    
    artist.penup()
    artist.goto(x, y)
    artist.pendown()
    
    shape_type = shapes[current_shape % len(shapes)]
    
    if shape_type == "square":
        for i in range(4):
            artist.forward(40)
            artist.right(90)
    elif shape_type == "triangle":
        for i in range(3):
            artist.forward(40)
            artist.right(120)
    elif shape_type == "circle":
        artist.circle(20)
    elif shape_type == "star":
        for i in range(5):
            artist.forward(40)
            artist.right(144)
    
    current_shape += 1
    print(f"Drew a {shape_type}!")

screen.onclick(draw_at_click)
print("Click to draw different shapes!")
print("Cycles through: square → triangle → circle → star")
turtle.done()
```

## 💥 Collision Detection Concepts!

In games, you need to check if things hit each other! This is called **collision detection**.

For turtles, we can check:
- Are two objects close to each other?
- Did the turtle hit the edge of the screen?
- Is the turtle at a specific location?

Let's learn how!


```python
# Collision detection: Catch the target!
import turtle
import random

screen = turtle.Screen()
screen.setup(width=600, height=600)

# Player turtle
player = turtle.Turtle()
player.shape("turtle")
player.color("green")
player.penup()

# Target
target = turtle.Turtle()
target.shape("circle")
target.color("red")
target.penup()
target.goto(random.randint(-250, 250), random.randint(-250, 250))

score = 0

def move_up():
    player.setheading(90)
    player.forward(20)
    check_collision()

def move_down():
    player.setheading(270)
    player.forward(20)
    check_collision()

def move_left():
    player.setheading(180)
    player.forward(20)
    check_collision()

def move_right():
    player.setheading(0)
    player.forward(20)
    check_collision()

def check_collision():
    global score
    # Check if player is close to target
    distance = player.distance(target)
    if distance < 20:
        score += 1
        print(f"Caught it! Score: {score}")
        # Move target to new location
        target.goto(random.randint(-250, 250), random.randint(-250, 250))

# Set up controls
screen.onkey(move_up, "Up")
screen.onkey(move_down, "Down")
screen.onkey(move_left, "Left")
screen.onkey(move_right, "Right")
screen.listen()

print("Use arrow keys to catch the red circle!")
print("Current score: 0")
turtle.done()
```

## 🎯 Practice Time: Boundary Detection!

Let's keep the turtle inside the screen!


```python
# Turtle can't leave the screen!
import turtle

screen = turtle.Screen()
screen.setup(width=600, height=600)
artist = turtle.Turtle()
artist.shape("turtle")

# Screen boundaries
MAX_X = 280
MAX_Y = 280

def move_forward():
    # Check if moving forward would go out of bounds
    artist.forward(20)
    
    # Get current position
    x = artist.xcor()
    y = artist.ycor()
    
    # Check boundaries and bounce back if needed
    if x > MAX_X:
        artist.goto(MAX_X, y)
        print("Hit right wall!")
    elif x < -MAX_X:
        artist.goto(-MAX_X, y)
        print("Hit left wall!")
    
    if y > MAX_Y:
        artist.goto(x, MAX_Y)
        print("Hit top wall!")
    elif y < -MAX_Y:
        artist.goto(x, -MAX_Y)
        print("Hit bottom wall!")

def turn_left():
    artist.left(45)

def turn_right():
    artist.right(45)

screen.onkey(move_forward, "Up")
screen.onkey(turn_left, "Left")
screen.onkey(turn_right, "Right")
screen.listen()

print("Use arrow keys to move. Can't leave the screen!")
turtle.done()
```

## ✏️ Mini-Project: Etch-A-Sketch!

Let's build a classic Etch-A-Sketch drawing toy! You can:
- Move in 4 directions with arrow keys
- Clear the screen
- Change pen color
- Lift the pen up/down


```python
# ETCH-A-SKETCH!
import turtle

screen = turtle.Screen()
screen.bgcolor("light gray")
screen.title("🖍️ Etch-A-Sketch 🖍️")

artist = turtle.Turtle()
artist.shape("circle")
artist.color("red")
artist.speed(0)

colors = ["red", "blue", "green", "purple", "orange", "black"]
color_index = 0

# Movement functions
def move_up():
    artist.setheading(90)
    artist.forward(10)

def move_down():
    artist.setheading(270)
    artist.forward(10)

def move_left():
    artist.setheading(180)
    artist.forward(10)

def move_right():
    artist.setheading(0)
    artist.forward(10)

# Control functions
def change_color():
    global color_index
    color_index = (color_index + 1) % len(colors)
    artist.color(colors[color_index])
    print(f"Color changed to: {colors[color_index]}")

def pen_up():
    artist.penup()
    print("Pen UP - not drawing")

def pen_down():
    artist.pendown()
    print("Pen DOWN - drawing")

def clear_screen():
    artist.clear()
    artist.penup()
    artist.home()
    artist.pendown()
    print("Screen cleared!")

def make_thicker():
    current = artist.pensize()
    artist.pensize(current + 1)
    print(f"Pen size: {current + 1}")

def make_thinner():
    current = artist.pensize()
    if current > 1:
        artist.pensize(current - 1)
        print(f"Pen size: {current - 1}")

# Bind keys
screen.onkey(move_up, "Up")
screen.onkey(move_down, "Down")
screen.onkey(move_left, "Left")
screen.onkey(move_right, "Right")
screen.onkey(change_color, "c")
screen.onkey(pen_up, "u")
screen.onkey(pen_down, "d")
screen.onkey(clear_screen, "space")
screen.onkey(make_thicker, "plus")
screen.onkey(make_thinner, "minus")
screen.listen()

print("=" * 40)
print("🖍️  ETCH-A-SKETCH CONTROLS  🖍️")
print("=" * 40)
print("Arrow Keys: Move and draw")
print("C: Change color")
print("U: Pen up (stop drawing)")
print("D: Pen down (start drawing)")
print("+/-: Change pen thickness")
print("Space: Clear screen")
print("=" * 40)

turtle.done()
```

## 🌟 Challenge Corner: Two-Player Pong Game!

Ready for the ultimate challenge? Build a two-player game where turtles compete!

Features:
- Two players with different controls
- A ball that moves and bounces
- Score keeping
- Boundaries

This is a simplified version - you can make it better!


```python
# TWO-PLAYER TAG GAME!
import turtle

screen = turtle.Screen()
screen.setup(width=600, height=600)
screen.bgcolor("light green")
screen.title("🏃 Two-Player Tag Game! 🏃")

# Player 1 (Red) - Arrow keys
player1 = turtle.Turtle()
player1.shape("turtle")
player1.color("red")
player1.penup()
player1.goto(-100, 0)

# Player 2 (Blue) - WASD keys
player2 = turtle.Turtle()
player2.shape("turtle")
player2.color("blue")
player2.penup()
player2.goto(100, 0)

# Score display
score_display = turtle.Turtle()
score_display.hideturtle()
score_display.penup()
score_display.goto(0, 250)

player1_score = 0
player2_score = 0

def update_score():
    score_display.clear()
    score_display.write(f"Red: {player1_score}  |  Blue: {player2_score}", 
                       align="center", font=("Arial", 16, "bold"))

# Player 1 controls (Arrow keys)
def p1_up():
    y = player1.ycor()
    if y < 270:
        player1.sety(y + 20)
    check_tag()

def p1_down():
    y = player1.ycor()
    if y > -270:
        player1.sety(y - 20)
    check_tag()

def p1_left():
    x = player1.xcor()
    if x > -270:
        player1.setx(x - 20)
    check_tag()

def p1_right():
    x = player1.xcor()
    if x < 270:
        player1.setx(x + 20)
    check_tag()

# Player 2 controls (WASD)
def p2_up():
    y = player2.ycor()
    if y < 270:
        player2.sety(y + 20)
    check_tag()

def p2_down():
    y = player2.ycor()
    if y > -270:
        player2.sety(y - 20)
    check_tag()

def p2_left():
    x = player2.xcor()
    if x > -270:
        player2.setx(x - 20)
    check_tag()

def p2_right():
    x = player2.xcor()
    if x < 270:
        player2.setx(x + 20)
    check_tag()

def check_tag():
    global player1_score, player2_score
    distance = player1.distance(player2)
    if distance < 20:
        # Red tagged Blue!
        player1_score += 1
        update_score()
        print(f"TAG! Red scores! Score is now Red:{player1_score} Blue:{player2_score}")
        # Reset positions
        player1.goto(-100, 0)
        player2.goto(100, 0)

# Bind keys
screen.onkey(p1_up, "Up")
screen.onkey(p1_down, "Down")
screen.onkey(p1_left, "Left")
screen.onkey(p1_right, "Right")

screen.onkey(p2_up, "w")
screen.onkey(p2_down, "s")
screen.onkey(p2_left, "a")
screen.onkey(p2_right, "d")

screen.listen()
update_score()

print("=" * 50)
print("🏃 TWO-PLAYER TAG GAME 🏃")
print("=" * 50)
print("RED PLAYER (arrows): Use Arrow Keys")
print("BLUE PLAYER (WASD): Use W, A, S, D Keys")
print("Goal: Red player tries to tag Blue player!")
print("First to 5 points wins!")
print("=" * 50)

turtle.done()
```
