## ⌨️ Keyboard Control!You can make your turtle move when you press keys on the keyboard! This is how games work!We'll use:- `onkey(function, "key")` - tells turtle what to do when a key is pressed- `listen()` - tells turtle to listen for keyboard

```python
# Random walker!import turtleimport randomartist = turtle.Turtle()artist.speed(0)screen = turtle.Screen()screen.bgcolor("white")colors = ["red", "blue", "green", "yellow", "purple", "orange", "pink"]for i in range(100):    # Pick random color    artist.color(random.choice(colors))        # Random size    size = random.randint(10, 100)    artist.circle(size)        # Random direction    artist.right(random.randint(0, 360))        # Random distance    artist.penup()    artist.forward(random.randint(10, 50))    artist.pendown()turtle.done()
```

## ✏️ Your Turn - Draw by Clicking!Let's make a program where you click the screen and the turtle draws there!

```python
# Click to draw!import turtlescreen = turtle.Screen()screen.bgcolor("lightyellow")artist = turtle.Turtle()artist.speed(0)artist.shape("circle")# Function that runs when you clickdef draw_dot(x, y):    artist.penup()    artist.goto(x, y)    artist.pendown()    artist.dot(20, "red")# Listen for clicksscreen.onclick(draw_dot)print("Click anywhere to draw dots!")turtle.done()
```

## 🌟 Ultimate Challenge - Drawing App!Create your own drawing app! Features to include:- Arrow keys to move the turtle- 'c' key to change color- 'u' key to pick pen up- 'd' key to put pen down- 's' key to stamp- Space bar to clear screenMake it YOUR way!

```python
# Your drawing app!import turtleimport randomscreen = turtle.Screen()screen.bgcolor("white")screen.title("Owen's Drawing App")artist = turtle.Turtle()artist.speed(0)colors = ["red", "blue", "green", "yellow", "purple", "orange", "pink", "black"]color_index = 0# Movement functionsdef forward():    artist.forward(20)def backward():    artist.backward(20)def left():    artist.left(45)def right():    artist.right(45)def change_color():    global color_index    color_index = (color_index + 1) % len(colors)    artist.color(colors[color_index])    print("Color:", colors[color_index])def pen_up():    artist.penup()    print("Pen up!")def pen_down():    artist.pendown()    print("Pen down!")def stamp_it():    artist.stamp()def clear_all():    artist.clear()    print("Screen cleared!")# Connect keysscreen.onkey(forward, "Up")screen.onkey(backward, "Down")screen.onkey(left, "Left")screen.onkey(right, "Right")screen.onkey(change_color, "c")screen.onkey(pen_up, "u")screen.onkey(pen_down, "d")screen.onkey(stamp_it, "s")screen.onkey(clear_all, "space")screen.listen()print("=== DRAWING APP ===")print("Arrow keys: Move")print("C: Change color")print("U/D: Pen up/down")print("S: Stamp")print("Space: Clear")turtle.done()
```

## 🖼️ Turtle Stamps!The turtle can leave copies of itself! Use `stamp()` to create a stamp. It's like taking a photo of the turtle!

```python
# Stamp trail!import turtleartist = turtle.Turtle()artist.shape("turtle")artist.color("green")artist.speed(0)# Make a trail of turtles!for i in range(20):    artist.stamp()    artist.forward(30)    artist.right(24)print("A spiral of turtles!")turtle.done()
```

## 🐢 Different Turtle Shapes!Your turtle doesn't have to look like a turtle! You can change its shape:- `"turtle"` - classic turtle- `"arrow"` - arrow pointer- `"circle"` - round dot- `"square"` - square- `"triangle"` - triangle- `"classic"` - classic arrow

```python
# Shape parade!import turtlescreen = turtle.Screen()screen.bgcolor("lightblue")shapes = ["turtle", "arrow", "circle", "square", "triangle"]colors = ["red", "blue", "green", "purple", "orange"]x_pos = -200for i in range(5):    t = turtle.Turtle()    t.shape(shapes[i])    t.color(colors[i])    t.penup()    t.goto(x_pos, 0)    t.stamp()    x_pos += 100turtle.done()
```

## 🌀 Amazing Spiral Art!Let's create mesmerizing spiral patterns that change colors!

```python
# Rainbow spiral!import turtleartist = turtle.Turtle()artist.speed(0)screen = turtle.Screen()screen.bgcolor("black")colors = ["red", "orange", "yellow", "green", "blue", "purple"]for i in range(360):    artist.pencolor(colors[i % 6])  # Cycle through colors    artist.width(i / 100 + 1)  # Line gets thicker    artist.forward(i)    artist.left(59)print("Spiral complete!")turtle.done()
```
