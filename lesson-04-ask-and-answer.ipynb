## 🔢 Input with Math

Here's something tricky: when you use `input()`, Python treats the answer as text (called a "string"). 

If you want to do math with the answer, you need to tell Python it's a number using `int()`:


```python
# Math with input!

print("Let's count your toys!")
cars = input("How many toy cars do you have? ")
cars = int(cars)  # Turn the text into a number!

dinosaurs = input("How many dinosaurs? ")
dinosaurs = int(dinosaurs)

total = cars + dinosaurs
print("\nYou have", total, "toys total!")

# Shortcut: you can do it in one line!
blocks = int(input("How many blocks? "))
print("Wow,", blocks, "blocks!")
```

## 🔢 Working with Decimal Numbers: float()

What if you need to work with decimal numbers like 3.14 or 7.5?

Use `float()` instead of `int()`!

- `int()` → whole numbers (1, 2, 3, 100)
- `float()` → decimal numbers (1.5, 3.14, 99.99)


```python
# Using float() for decimal numbers
print("💰 MONEY CALCULATOR 💰\n")

allowance = float(input("How much allowance do you get? $"))
savings = float(input("How much have you saved? $"))

total_money = allowance + savings
print("\nTotal money: $", total_money)

print("\n---")

# Calculate a price
price = float(input("\nHow much does the toy cost? $"))
if price <= total_money:
    money_left = total_money - price
    print("You can afford it! You'll have $", money_left, "left.")
else:
    print("You need to save more money!")
```

## ✨ String Methods: Cleaning Up Input!

Sometimes people type in weird ways:
- "OWEN" (all caps)
- "  sarah  " (extra spaces)
- "Alex" (mixed case)

Python has special **string methods** to clean up text!

- `.lower()` → makes everything lowercase: "HELLO" → "hello"
- `.upper()` → makes everything UPPERCASE: "hello" → "HELLO"
- `.strip()` → removes extra spaces: "  hi  " → "hi"
- `.title()` → Capitalizes Each Word: "hello world" → "Hello World"


```python
# String methods in action!
print("STRING METHODS DEMO\n")

name = input("What's your name? ")

# Try different methods
print("\nOriginal:", name)
print("Lowercase:", name.lower())
print("Uppercase:", name.upper())
print("Title Case:", name.title())
print("Stripped:", name.strip())  # Removes spaces at start/end

print("\n---")

# Practical use: making input consistent
favorite_color = input("\nWhat's your favorite color? ").lower().strip()
print("You said:", favorite_color)

# Now we can check easily!
if favorite_color == "blue":
    print("Blue is a cool color!")
elif favorite_color == "red":
    print("Red is a warm color!")
else:
    print(favorite_color.title(), "is a great choice!")
```

## 🛡️ Input Validation: Checking if Input Makes Sense

Sometimes people type things that don't make sense! Like:
- Entering "banana" when you ask for a number
- Leaving an answer blank
- Entering a negative age

We can check if input is reasonable before using it!


```python
# Input validation examples
print("🛡️ SMART INPUT CHECKER 🛡️\n")

# Check if a string is not empty
name = input("What's your name? ").strip()
if name == "":
    print("❌ You didn't enter a name!")
else:
    print(f"✅ Hello, {name}!")

print("\n---")

# Check if input is a reasonable number
age_input = input("How old are you? ")
try:
    age = int(age_input)
    if age < 0:
        print("❌ Age can't be negative!")
    elif age > 150:
        print("❌ That's too old to be real!")
    else:
        print(f"✅ You are {age} years old.")
except:
    print("❌ That's not a valid number!")

print("\n---")

# Check if answer is in a list of valid choices
print("Choose a color: red, blue, or green")
color = input("Your choice: ").lower().strip()
valid_colors = ["red", "blue", "green"]
if color in valid_colors:
    print(f"✅ {color.title()} is a great choice!")
else:
    print(f"❌ {color} is not one of the choices!")
```

```python
# RESTAURANT ORDER SYSTEM
print("🍔 PYTHON'S BURGER SHACK 🍔")
print("=" * 40)
print("        MENU")
print("=" * 40)
print("Burger.............$5.00")
print("Fries..............$3.00")
print("Soda...............$2.00")
print("Ice Cream..........$4.00")
print("=" * 40)

# Get customer info
customer_name = input("\nWhat's your name? ").strip().title()
print(f"\nHello {customer_name}! Let's take your order.\n")

# Take the order
burger_qty = int(input("How many burgers? "))
fries_qty = int(input("How many orders of fries? "))
soda_qty = int(input("How many sodas? "))
icecream_qty = int(input("How many ice creams? "))

# Calculate costs
burger_cost = burger_qty * 5.00
fries_cost = fries_qty * 3.00
soda_cost = soda_qty * 2.00
icecream_cost = icecream_qty * 4.00

subtotal = burger_cost + fries_cost + soda_cost + icecream_cost
tax = subtotal * 0.08  # 8% tax
total = subtotal + tax

# Show the receipt
print("\n" + "=" * 40)
print(f"      RECEIPT FOR {customer_name.upper()}")
print("=" * 40)
if burger_qty > 0:
    print(f"Burgers x{burger_qty}........${burger_cost:.2f}")
if fries_qty > 0:
    print(f"Fries x{fries_qty}..........${fries_cost:.2f}")
if soda_qty > 0:
    print(f"Sodas x{soda_qty}..........${soda_cost:.2f}")
if icecream_qty > 0:
    print(f"Ice Cream x{icecream_qty}...${icecream_cost:.2f}")
print("-" * 40)
print(f"Subtotal..........${subtotal:.2f}")
print(f"Tax (8%)..........${tax:.2f}")
print(f"TOTAL.............${total:.2f}")
print("=" * 40)
print(f"\nThank you, {customer_name}! Enjoy your meal! 🎉")
```

## 🌟 Challenge Corner: Simple Chatbot!

Ready for the ultimate challenge? Build a chatbot that can have a conversation with you!

It will:
- Greet you
- Ask several questions
- Respond differently based on your answers
- Remember what you said
- Say goodbye nicely


```python
# SIMPLE CHATBOT
print("🤖 CHATBOT 3000 🤖")
print("=" * 50)
print("Hello! I'm ChatBot 3000, your friendly AI assistant!")
print("=" * 50)

# Introduction
name = input("\nWhat's your name? ").strip().title()
print(f"\nNice to meet you, {name}! I love that name!")

# Ask about their day
print(f"\nSo {name}, tell me about yourself...")
age = input("How old are you? ")
print(f"Wow, {age}! That's a fun age!")

# Favorite things
fav_color = input("\nWhat's your favorite color? ").lower().strip()
if fav_color == "blue":
    print("Blue is my favorite too! Great minds think alike!")
elif fav_color == "red":
    print("Red is so energetic! I like it!")
elif fav_color == "green":
    print("Green is the color of nature. Beautiful choice!")
else:
    print(f"{fav_color.title()} sounds wonderful!")

# Hobbies
hobby = input("\nWhat do you like to do for fun? ").strip()
print(f"{hobby.title()} sounds awesome! I wish I could try that!")

# Pet question
has_pet = input("\nDo you have any pets? (yes/no): ").lower().strip()
if has_pet == "yes" or has_pet == "y":
    pet_type = input("What kind of pet? ")
    pet_name = input("What's your pet's name? ").strip().title()
    print(f"{pet_name} the {pet_type}! That's adorable! 🐾")
else:
    print("Maybe you'll get a pet someday!")

# School
grade = input("\nWhat grade are you in? ")
subject = input("What's your favorite subject in school? ").strip()
print(f"{subject.title()} in grade {grade}! You must be really smart!")

# Dream question
print(f"\nOne more question, {name}...")
dream = input("What do you want to be when you grow up? ").strip()
print(f"A {dream}! That's an amazing dream!")
print(f"I believe in you, {name}! You can do anything!")

# Summary and goodbye
print("\n" + "=" * 50)
print("🌟 CONVERSATION SUMMARY 🌟")
print("=" * 50)
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Favorite Color: {fav_color.title()}")
print(f"Hobby: {hobby.title()}")
if has_pet == "yes" or has_pet == "y":
    print(f"Pet: {pet_name} the {pet_type}")
print(f"Grade: {grade}")
print(f"Favorite Subject: {subject.title()}")
print(f"Dream Job: {dream.title()}")
print("=" * 50)

print(f"\nIt was great talking to you, {name}!")
print("I hope we can chat again soon! Goodbye! 👋")
print("=" * 50)
```
