## 👂 The input() FunctionThe `input()` function is like giving Python ears! It lets Python:1. Ask you a question2. Wait for you to type an answer3. Remember what you saidLet's try it!

## 🧮 Build a Calculator!Let's combine what we learned to make a simple calculator that asks YOU what to calculate!

```python
# Your own calculator!print("=== OWEN'S CALCULATOR ===")num1 = int(input("Enter first number: "))num2 = int(input("Enter second number: "))print("\n--- Results ---")print(num1, "+", num2, "=", num1 + num2)print(num1, "-", num2, "=", num1 - num2)print(num1, "*", num2, "=", num1 * num2)print(num1, "/", num2, "=", num1 / num2)
```

## 🔢 Input with MathHere's something tricky: when you use `input()`, Python treats the answer as text (called a "string"). If you want to do math with the answer, you need to tell Python it's a number using `int()`:

```python
# Math with input!print("Let's count your toys!")cars = input("How many toy cars do you have? ")cars = int(cars)  # Turn the text into a number!dinosaurs = input("How many dinosaurs? ")dinosaurs = int(dinosaurs)total = cars + dinosaursprint("\nYou have", total, "toys total!")# Shortcut: you can do it in one line!blocks = int(input("How many blocks? "))print("Wow,", blocks, "blocks!")
```
