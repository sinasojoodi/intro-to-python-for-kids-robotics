## 📝 Secret Notes: Comments!

Sometimes you want to write notes in your code that Python will ignore. These are called **comments**!

Comments start with a `#` symbol. Python skips over them - they're just for humans to read!


```python
# This is a comment - Python ignores it!
print("But Python reads this!")

# You can use comments to explain your code
print("Hello!")  # You can even put comments at the end of a line!

# Comments are great for leaving notes to yourself
# or explaining what your code does
```

## ✨ Magic Escape Characters!

Sometimes you want to do special things in your text, like make a new line or add a tab space. Python has **escape characters** for this!

- `\n` means "new line" (go to the next line)
- `\t` means "tab" (add a big space)

The backslash `\` tells Python "the next character is special!"


```python
# Using \n to make new lines
print("Line 1\nLine 2\nLine 3")

print("\n---\n")  # Empty lines to separate!

# Using \t to make tabs (big spaces)
print("Name:\tOwen")
print("Age:\t7")
print("Grade:\t2nd")

print("\n---\n")

# You can use both together!
print("Hello!\n\tThis line is indented!\n\t\tThis one is indented even more!")
```
