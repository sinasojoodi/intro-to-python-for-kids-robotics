## 🎨 Adding Colors!

You can change the color of your turtle's pen and even the turtle itself!

**Color Commands:**
- `pencolor("color_name")` - changes pen color
- `fillcolor("color_name")` - changes fill color
- `color("color_name")` - changes both pen and turtle color

**Popular colors:** red, blue, green, yellow, purple, orange, pink, brown, black


```python
# Rainbow circles!
import turtle

artist = turtle.Turtle()
artist.speed(8)

colors = ["red", "orange", "yellow", "green", "blue", "purple"]

for color in colors:
    artist.color(color)
    artist.circle(50)
    artist.right(60)  # Turn a bit for next circle

print("Rainbow pattern complete!")
turtle.done()
```

## ⭐ Draw a Star!

```python
# Drawing a star!
import turtle

artist = turtle.Turtle()
artist.speed(3)
artist.color("gold")

# Five-pointed star
artist.begin_fill()
for i in range(5):
    artist.forward(100)
    artist.right(144)  # Special angle for stars!
artist.end_fill()

print("You made a star!")
turtle.done()
```

## 🌟 Ultimate Challenge - Flower Garden!

Create a colorful flower garden using:
- Loops to draw multiple flowers
- Different colors for each flower
- Filled circles for petals

Make it beautiful!


```python
# Flower garden!
import turtle

artist = turtle.Turtle()
artist.speed(0)  # Super fast!

# Example: one flower
def draw_flower(x, y, color):
    artist.penup()
    artist.goto(x, y)
    artist.pendown()
    artist.color(color)
    
    # Draw 6 petals
    for i in range(6):
        artist.begin_fill()
        artist.circle(30)
        artist.end_fill()
        artist.left(60)

# Draw multiple flowers!
draw_flower(0, 0, "red")
draw_flower(100, 50, "yellow")
draw_flower(-100, 30, "pink")
# Add more!

turtle.done()
```

## 🎨 Filling Shapes with Color!

You can fill shapes with color! Here's how:
1. Call `begin_fill()` before drawing
2. Draw your shape
3. Call `end_fill()` when done

Watch the magic happen!


```python
# Filled shapes!
import turtle

artist = turtle.Turtle()
artist.speed(2)

# Yellow filled circle (like the sun!)
artist.color("yellow")
artist.begin_fill()
artist.circle(60)
artist.end_fill()

# Green filled triangle (like a tree!)
artist.penup()
artist.goto(150, 0)
artist.pendown()

artist.color("green")
artist.begin_fill()
artist.forward(100)
artist.left(120)
artist.forward(100)
artist.left(120)
artist.forward(100)
artist.end_fill()

turtle.done()
```

## 🔁 Magic Loops!

Remember when we drew a square? We had to write the same code 4 times! That's boring!

**Loops** let us repeat code automatically! Instead of writing it 4 times, we just say "do this 4 times!"

It's like a magic repeat button!


```python
# Square the EASY way with a loop!
import turtle

artist = turtle.Turtle()
artist.speed(2)
artist.color("purple")

# Old way (boring):
# artist.forward(100)
# artist.right(90)
# artist.forward(100)
# artist.right(90)
# artist.forward(100)
# artist.right(90)
# artist.forward(100)
# artist.right(90)

# New way with a loop (cool!):
for i in range(4):  # Do this 4 times!
    artist.forward(100)
    artist.right(90)

print("Loop is magic!")
turtle.done()
```

## 🔷 Drawing Any Shape with Loops!

With loops, you can draw shapes with ANY number of sides!
- Triangle = 3 sides
- Square = 4 sides
- Pentagon = 5 sides
- Hexagon = 6 sides
- Octagon = 8 sides!

The trick: divide 360 by the number of sides to get the turn angle!


```python
# Draw different shapes!
import turtle

artist = turtle.Turtle()
artist.speed(5)

# Hexagon (6 sides)
artist.color("orange")
for i in range(6):
    artist.forward(80)
    artist.right(60)  # 360 / 6 = 60

# Octagon (8 sides)
artist.penup()
artist.goto(150, 0)
artist.pendown()

artist.color("red")
for i in range(8):
    artist.forward(50)
    artist.right(45)  # 360 / 8 = 45

turtle.done()
```

## 🎨 Color Mixing Concepts!

Did you know you can create ANY color using RGB (Red, Green, Blue)?

Each color has three values from 0 to 255:
- (255, 0, 0) = Red
- (0, 255, 0) = Green  
- (0, 0, 255) = Blue
- (255, 255, 0) = Yellow (red + green)
- (255, 0, 255) = Magenta (red + blue)
- (0, 255, 255) = Cyan (green + blue)
- (255, 255, 255) = White (all colors!)
- (0, 0, 0) = Black (no color)

To use RGB, you need to set the color mode first!


```python
# Using RGB colors!
import turtle

artist = turtle.Turtle()
artist.speed(5)

# Enable RGB color mode (0-255)
turtle.colormode(255)

# Pure red
artist.color(255, 0, 0)
artist.begin_fill()
artist.circle(30)
artist.end_fill()

# Pure green
artist.penup()
artist.goto(80, 0)
artist.pendown()
artist.color(0, 255, 0)
artist.begin_fill()
artist.circle(30)
artist.end_fill()

# Pure blue
artist.penup()
artist.goto(160, 0)
artist.pendown()
artist.color(0, 0, 255)
artist.begin_fill()
artist.circle(30)
artist.end_fill()

# Custom purple (mixing red and blue)
artist.penup()
artist.goto(40, 80)
artist.pendown()
artist.color(200, 0, 200)
artist.begin_fill()
artist.circle(30)
artist.end_fill()

turtle.done()
```

## 🌈 Gradient Patterns!

A gradient is when colors slowly change from one to another. Let's create beautiful gradients!

We can make a color gradually change by slowly increasing or decreasing the RGB values!


```python
# Red to blue gradient!
import turtle

artist = turtle.Turtle()
artist.speed(0)
turtle.colormode(255)

# Start at the left
artist.penup()
artist.goto(-200, 0)
artist.pendown()

# Draw 20 circles, gradually changing from red to blue
for i in range(20):
    # Calculate color values
    red_amount = 255 - (i * 12)  # Red decreases
    blue_amount = i * 12          # Blue increases
    
    artist.color(red_amount, 0, blue_amount)
    artist.begin_fill()
    artist.circle(20)
    artist.end_fill()
    
    # Move to next position
    artist.penup()
    artist.forward(20)
    artist.pendown()

print("Beautiful gradient!")
turtle.done()
```

## 🎨 Shape Gallery!

Let's create a gallery of different shapes, each with its own color!


```python
# SHAPE GALLERY!
import turtle

artist = turtle.Turtle()
artist.speed(8)

# Gallery setup
shapes_data = [
    (3, "red", -200, 100, "Triangle"),      # (sides, color, x, y, name)
    (4, "blue", -100, 100, "Square"),
    (5, "green", 0, 100, "Pentagon"),
    (6, "orange", 100, 100, "Hexagon"),
    (8, "purple", -150, -50, "Octagon"),
    (0, "pink", -50, -50, "Circle"),        # 0 sides = circle
    (12, "brown", 50, -50, "Dodecagon")
]

# Draw each shape
for shape_info in shapes_data:
    sides, color, x, y, name = shape_info
    
    # Move to position
    artist.penup()
    artist.goto(x, y)
    artist.pendown()
    
    # Set color
    artist.color(color)
    artist.begin_fill()
    
    # Draw the shape
    if sides == 0:  # Circle
        artist.circle(30)
    else:
        angle = 360 / sides
        for i in range(sides):
            artist.forward(50)
            artist.right(angle)
    
    artist.end_fill()
    
    # Label it
    artist.penup()
    artist.goto(x, y - 50)
    artist.write(name, align="center", font=("Arial", 10, "normal"))

print("Shape gallery complete!")
turtle.done()
```

## 🔁 Nested Loops: Patterns Within Patterns!

A **nested loop** is a loop inside another loop! This lets you create complex patterns!

Think of it like:
- Outer loop: "Do this 5 times"
- Inner loop: "For each of those times, do THIS 4 times"

It's super powerful!


```python
# Grid of squares using nested loops!
import turtle

artist = turtle.Turtle()
artist.speed(0)

colors = ["red", "orange", "yellow", "green", "blue", "purple"]

# Outer loop: rows
for row in range(4):
    # Inner loop: columns
    for col in range(6):
        # Move to position
        artist.penup()
        artist.goto(col * 60 - 150, row * 60 - 100)
        artist.pendown()
        
        # Pick a color
        artist.color(colors[(row + col) % 6])
        artist.begin_fill()
        
        # Draw square
        for side in range(4):
            artist.forward(50)
            artist.right(90)
        
        artist.end_fill()

print("Grid pattern complete!")
turtle.done()
```

## 🎬 Simple Animation: Growing Circles!

Let's make our art move! We can create simple animations by drawing, clearing, and redrawing!


```python
# Growing circle animation!
import turtle
import time

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)

colors = ["red", "orange", "yellow", "green", "blue", "purple"]

# Draw circles that grow
for i in range(50):
    color = colors[i % 6]  # Cycle through colors
    artist.color(color)
    artist.circle(i * 5)
    artist.right(10)  # Turn a bit for spiral effect

print("Animation complete!")
turtle.done()
```

```python
# Spinning square animation!
import turtle

screen = turtle.Screen()
artist = turtle.Turtle()
artist.speed(0)
turtle.colormode(255)

# Draw 36 squares, each rotated 10 degrees
for i in range(36):
    # Calculate color (gradient from red to blue)
    red = 255 - (i * 7)
    blue = i * 7
    
    artist.color(red, 0, blue)
    
    # Draw square
    for side in range(4):
        artist.forward(100)
        artist.right(90)
    
    # Rotate for next square
    artist.right(10)

print("Spinning squares complete!")
turtle.done()
```

## 🌟 Challenge Corner: Create a Mandala!

A **mandala** is a beautiful circular pattern! It's the ultimate art challenge!

You'll need:
- Loops to repeat patterns
- Colors that change
- Rotation to make it circular
- Nested loops for complexity

Let's build one step by step!


```python
# SIMPLE MANDALA
import turtle

artist = turtle.Turtle()
artist.speed(0)
turtle.bgcolor("black")  # Black background looks awesome!

colors = ["red", "orange", "yellow", "green", "blue", "purple", "pink"]

# Draw 36 circles in a circle pattern
for i in range(36):
    artist.color(colors[i % 7])
    artist.circle(100)
    artist.right(10)  # Rotate 10 degrees each time

print("Simple mandala complete!")
turtle.done()
```

```python
# ADVANCED MANDALA - Petals!
import turtle

artist = turtle.Turtle()
artist.speed(0)
turtle.bgcolor("midnight blue")
turtle.colormode(255)

# Draw 36 petals around the center
for i in range(36):
    # Calculate gradient color
    red = (i * 7) % 255
    green = 100
    blue = 255 - ((i * 7) % 255)
    
    artist.color(red, green, blue)
    artist.begin_fill()
    
    # Draw one petal (using circles)
    artist.circle(80, 60)  # Draw 60 degrees of a circle
    artist.left(120)
    artist.circle(80, 60)
    artist.left(120)
    
    artist.end_fill()
    
    # Rotate to next petal position
    artist.left(10)

print("Beautiful mandala!")
turtle.done()
```
