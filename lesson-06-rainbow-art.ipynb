## 🎨 Adding Colors!You can change the color of your turtle's pen and even the turtle itself!**Color Commands:**- `pencolor("color_name")` - changes pen color- `fillcolor("color_name")` - changes fill color- `color("color_name")` - changes both pen and turtle color**Popular colors:** red, blue, green, yellow, purple, orange, pink, brown, black

```python
# Rainbow circles!import turtleartist = turtle.Turtle()artist.speed(8)colors = ["red", "orange", "yellow", "green", "blue", "purple"]for color in colors:    artist.color(color)    artist.circle(50)    artist.right(60)  # Turn a bit for next circleprint("Rainbow pattern complete!")turtle.done()
```

## ⭐ Draw a Star!Want to draw a star? It's just a loop with the right angles!

```python
# Drawing a star!import turtleartist = turtle.Turtle()artist.speed(3)artist.color("gold")# Five-pointed starartist.begin_fill()for i in range(5):    artist.forward(100)    artist.right(144)  # Special angle for stars!artist.end_fill()print("You made a star!")turtle.done()
```

## 🌟 Ultimate Challenge - Flower Garden!Create a colorful flower garden using:- Loops to draw multiple flowers- Different colors for each flower- Filled circles for petalsMake it beautiful!

```python
# Flower garden!import turtleartist = turtle.Turtle()artist.speed(0)  # Super fast!# Example: one flowerdef draw_flower(x, y, color):    artist.penup()    artist.goto(x, y)    artist.pendown()    artist.color(color)        # Draw 6 petals    for i in range(6):        artist.begin_fill()        artist.circle(30)        artist.end_fill()        artist.left(60)# Draw multiple flowers!draw_flower(0, 0, "red")draw_flower(100, 50, "yellow")draw_flower(-100, 30, "pink")# Add more!turtle.done()
```

## 🎨 Filling Shapes with Color!You can fill shapes with color! Here's how:1. Call `begin_fill()` before drawing2. Draw your shape3. Call `end_fill()` when doneWatch the magic happen!

```python
# Filled shapes!import turtleartist = turtle.Turtle()artist.speed(2)# Yellow filled circle (like the sun!)artist.color("yellow")artist.begin_fill()artist.circle(60)artist.end_fill()# Green filled triangle (like a tree!)artist.penup()artist.goto(150, 0)artist.pendown()artist.color("green")artist.begin_fill()artist.forward(100)artist.left(120)artist.forward(100)artist.left(120)artist.forward(100)artist.end_fill()turtle.done()
```

## 🔁 Magic Loops!Remember when we drew a square? We had to write the same code 4 times! That's boring!**Loops** let us repeat code automatically! Instead of writing it 4 times, we just say "do this 4 times!"It's like a magic repeat button!

```python
# Square the EASY way with a loop!import turtleartist = turtle.Turtle()artist.speed(2)artist.color("purple")# Old way (boring):# artist.forward(100)# artist.right(90)# artist.forward(100)# artist.right(90)# artist.forward(100)# artist.right(90)# artist.forward(100)# artist.right(90)# New way with a loop (cool!):for i in range(4):  # Do this 4 times!    artist.forward(100)    artist.right(90)print("Loop is magic!")turtle.done()
```

## 🔷 Drawing Any Shape with Loops!With loops, you can draw shapes with ANY number of sides!- Triangle = 3 sides- Square = 4 sides- Pentagon = 5 sides- Hexagon = 6 sides- Octagon = 8 sides!The trick: divide 360 by the number of sides to get the turn angle!

```python
# Draw different shapes!import turtleartist = turtle.Turtle()artist.speed(5)# Hexagon (6 sides)artist.color("orange")for i in range(6):    artist.forward(80)    artist.right(60)  # 360 / 6 = 60# Octagon (8 sides)artist.penup()artist.goto(150, 0)artist.pendown()artist.color("red")for i in range(8):    artist.forward(50)    artist.right(45)  # 360 / 8 = 45turtle.done()
```
