## ⚖️ Comparison OperatorsTo check conditions, we use special symbols:- `==` means "is equal to" (not just `=`!)- `!=` means "is not equal to"- `>` means "greater than"- `<` means "less than"- `>=` means "greater than or equal to"- `<=` means "less than or equal to"These help Python make decisions!

````python
# Testing conditions!score = 95if score >= 90:    print("Amazing! You got an A!")temperature = 75if temperature > 80:    print("It's hot outside!")if temperature <= 80:    print("Nice temperature!")name = "Owen"if name == "Owen":    print("Hey, that's you!")
````

## 🔄 The else StatementWhat if the condition is False? We can tell Python what to do instead using `else`:```if condition:    do this if Trueelse:    do this if False```It's like saying "if this, do that, otherwise do something else!"

````python
# Using if-else!weather = "sunny"if weather == "rainy":    print("Bring an umbrella!")else:    print("No umbrella needed!")print("\n--- Snack Time ---")hungry = Trueif hungry:    print("Let's eat a snack!")else:    print("No snack needed right now")# Try changing weather to "rainy" or hungry to False!
````

## 🎯 The elif StatementWhat if you have MORE than 2 choices? Use `elif` (which means "else if"):```if condition1:    do thiselif condition2:    do this insteadelse:    do this if nothing else matched```

````python
# Multiple choices with elif!score = 85if score >= 90:    print("Grade: A - Excellent!")elif score >= 80:    print("Grade: B - Great job!")elif score >= 70:    print("Grade: C - Good!")else:    print("Keep practicing!")print("\n--- Weather Check ---")temperature = 70if temperature > 85:    print("It's HOT! 🔥")elif temperature > 70:    print("It's warm! ☀️")elif temperature > 50:    print("It's cool! 🍂")else:    print("It's cold! ❄️")
````
