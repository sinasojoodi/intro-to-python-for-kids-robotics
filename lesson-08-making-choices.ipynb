## ⚖️ Comparison Operators

To check conditions, we use special symbols:

- `==` means "is equal to" (not just `=`!)
- `!=` means "is not equal to"
- `>` means "greater than"
- `<` means "less than"
- `>=` means "greater than or equal to"
- `<=` means "less than or equal to"

These help Python make decisions!


````python
# Testing conditions!

score = 95

if score >= 90:
    print("Amazing! You got an A!")

temperature = 75

if temperature > 80:
    print("It's hot outside!")

if temperature <= 80:
    print("Nice temperature!")

name = "Owen"

if name == "Owen":
    print("Hey, that's you!")
````

## 🔄 The else Statement

What if the condition is False? We can tell Python what to do instead using `else`:

```
if condition:
    do this if True
else:
    do this if False
```

It's like saying "if this, do that, otherwise do something else!"


````python
# Using if-else!

weather = "sunny"

if weather == "rainy":
    print("Bring an umbrella!")
else:
    print("No umbrella needed!")

print("\n--- Snack Time ---")
hungry = True

if hungry:
    print("Let's eat a snack!")
else:
    print("No snack needed right now")

# Try changing weather to "rainy" or hungry to False!
````

## 🎯 The elif Statement

What if you have MORE than 2 choices? Use `elif` (which means "else if"):

```
if condition1:
    do this
elif condition2:
    do this instead
else:
    do this if nothing else matched
```


````python
# Multiple choices with elif!

score = 85

if score >= 90:
    print("Grade: A - Excellent!")
elif score >= 80:
    print("Grade: B - Great job!")
elif score >= 70:
    print("Grade: C - Good!")
else:
    print("Keep practicing!")

print("\n--- Weather Check ---")
temperature = 70

if temperature > 85:
    print("It's HOT! 🔥")
elif temperature > 70:
    print("It's warm! ☀️")
elif temperature > 50:
    print("It's cool! 🍂")
else:
    print("It's cold! ❄️")
````

## 🔗 Boolean Operators: and, or, not!

Sometimes you need to check MULTIPLE things at once! That's where Boolean operators help:

- `and` - both conditions must be True
- `or` - at least one condition must be True
- `not` - flips True to False (and vice versa)

These let you make complex decisions!


````python
# Using 'and' - both must be true
age = 10
has_ticket = True

if age >= 7 and has_ticket:
    print("You can ride the roller coaster!")
else:
    print("Sorry, you need to be 7+ AND have a ticket")

print("\n---")

# Using 'or' - at least one must be true
is_weekend = True
is_holiday = False

if is_weekend or is_holiday:
    print("No school today! Yay!")
else:
    print("Time for school!")

print("\n---")

# Using 'not' - flips the condition
raining = False

if not raining:
    print("Great! We can play outside!")
else:
    print("Let's play inside today")

print("\n---")

# Combining them!
temperature = 75
sunny = True

if temperature > 70 and sunny and not raining:
    print("Perfect beach weather! 🏖️")
else:
    print("Maybe not beach weather today")
````

## 🏗️ Nested if Statements!

You can put if statements INSIDE other if statements! This is called **nesting**.

It's like making decisions step by step:
1. First, check THIS
2. If that's true, THEN check THIS OTHER THING
3. And so on...


````python
# Nested if example: Amusement park ride checker
print("🎢 RIDE ELIGIBILITY CHECKER 🎢\n")

print("How tall are you in cm? ")
height = int(input("How tall are you in cm? "))

if height >= 120:
    print("✅ You're tall enough!")
    
    # Now check another condition (nested)
    print("How old are you? ")
    age = int(input("How old are you? "))
    
    if age >= 10:
        print("✅ You can ride alone!")
    else:
        print("⚠️ You need an adult with you")
else:
    print("❌ Sorry, you must be at least 120cm tall")
    
print("\n---")

# Another example: Game character creator
print("⚔️ CHARACTER CREATOR ⚔️\n")

print("Choose class (warrior/mage/rogue): ")
character_type = input("Choose class (warrior/mage/rogue): ").lower()

if character_type == "warrior":
    print("You chose Warrior!")
    
    print("Choose weapon (sword/axe): ")
    weapon = input("Choose weapon (sword/axe): ").lower()
    
    if weapon == "sword":
        print("🗡️ You are a Sword Warrior! High speed!")
    elif weapon == "axe":
        print("🪓 You are an Axe Warrior! High power!")
    else:
        print("You get a basic weapon")
        
elif character_type == "mage":
    print("You chose Mage!")
    
    print("Choose magic (fire/ice): ")
    magic = input("Choose magic (fire/ice): ").lower()
    
    if magic == "fire":
        print("🔥 You are a Fire Mage! Burn enemies!")
    elif magic == "ice":
        print("❄️ You are an Ice Mage! Freeze enemies!")
    else:
        print("You get basic magic")
        
elif character_type == "rogue":
    print("🗡️ You chose Rogue! Stealth master!")
    
else:
    print("Unknown class. You're an Adventurer!")

print("\nCharacter created!")
````

## 🔬 Mini-Project: Animal Classifier!

Let's build a simple animal classifier using if-elif-else!


````python
# ANIMAL CLASSIFIER!
print("=" * 50)
print("🐾 ANIMAL CLASSIFIER 3000 🐾")
print("=" * 50)
print("\nAnswer these questions to classify an animal!\n")

# Ask questions
print("Does it have fur? (yes/no): ")
has_fur = input("Does it have fur? (yes/no): ").lower() == "yes"
print("Can it fly? (yes/no): ")
can_fly = input("Can it fly? (yes/no): ").lower() == "yes"
print("Does it live in water? (yes/no): ")
lives_in_water = input("Does it live in water? (yes/no): ").lower() == "yes"
print("Does it have scales? (yes/no): ")
has_scales = input("Does it have scales? (yes/no): ").lower() == "yes"
print("Does it lay eggs? (yes/no): ")
lays_eggs = input("Does it lay eggs? (yes/no): ").lower() == "yes"

print("\n" + "=" * 50)
print("CLASSIFICATION RESULT:")
print("=" * 50)

# Classification logic
if has_fur and not lives_in_water:
    if can_fly:
        print("🦇 This is probably a BAT!")
    else:
        print("🐕 This is probably a MAMMAL (like a dog, cat, or bear)!")
        
elif can_fly and not has_fur:
    if has_scales:
        print("🐉 This could be a mythical DRAGON!")
    else:
        print("🦅 This is probably a BIRD!")
        
elif lives_in_water:
    if has_scales:
        print("🐟 This is probably a FISH!")
    elif has_fur:
        print("🦭 This is probably a MARINE MAMMAL (like a seal or otter)!")
    else:
        print("🐙 This could be an INVERTEBRATE (like an octopus or jellyfish)!")
        
elif has_scales and not lives_in_water:
    if can_fly:
        print("🐉 Might be a mythical flying reptile!")
    elif lays_eggs:
        print("🦎 This is probably a REPTILE (like a lizard or snake)!")
    else:
        print("🐍 This might be a SNAKE!")
        
elif lays_eggs and not can_fly:
    print("🐢 This could be an AMPHIBIAN or land-based egg-layer!")
    
else:
    print("🤔 Hmm, that's a tricky one! Could be many things!")

print("=" * 50)
````
