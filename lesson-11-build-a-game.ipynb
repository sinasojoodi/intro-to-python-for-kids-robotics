```python
# Dice Rolling Game!
import random

print("=== DICE DUEL ===")
print("Roll higher than the computer to win!\n")

player_score = 0
computer_score = 0
rounds = 3

for round_num in range(1, rounds + 1):
    print(f"--- Round {round_num} ---")
    input("Press Enter to roll your dice...")
    
    player_roll = random.randint(1, 6)
    computer_roll = random.randint(1, 6)
    
    print(f"You rolled: 🎲 {player_roll}")
    print(f"Computer rolled: 🎲 {computer_roll}")
    
    if player_roll > computer_roll:
        print("✅ You win this round!")
        player_score += 1
    elif computer_roll > player_roll:
        print("❌ Computer wins this round!")
        computer_score += 1
    else:
        print("🤝 It's a tie!")
    
    print(f"Score - You: {player_score}, Computer: {computer_score}\n")

print("=== FINAL RESULT ===")
if player_score > computer_score:
    print("🎉 YOU WIN THE GAME!")
elif computer_score > player_score:
    print("😢 Computer wins!")
else:
    print("🤝 It's a tie game!")
```

## 🔢 Game 2: Advanced Number Guessing

Remember the number guessing game? Let's make it MUCH better!


```python
# Advanced Number Guessing Game!
import random

print("=== ULTIMATE NUMBER GUESSER ===")
print("I'm thinking of a number between 1 and 50!")
print("You have 7 tries to guess it!\n")

secret_number = random.randint(1, 50)
max_tries = 7
tries_used = 0

while tries_used < max_tries:
    tries_left = max_tries - tries_used
    print(f"Tries left: {tries_left}")
    
    print("Your guess: ")
    guess = int(input("Your guess: "))
    tries_used += 1
    
    if guess == secret_number:
        print(f"\n🎉 AMAZING! You got it in {tries_used} tries!")
        break
    elif guess < secret_number:
        difference = secret_number - guess
        if difference <= 5:
            print("🔥 Very close! Go HIGHER!")
        elif difference <= 10:
            print("📈 Close! Go higher!")
        else:
            print("📈 Too low! Go much higher!")
    elif guess > secret_number:
        difference = guess - secret_number
        if difference <= 5:
            print("🔥 Very close! Go LOWER!")
        elif difference <= 10:
            print("📉 Close! Go lower!")
        else:
            print("📉 Too high! Go much lower!")
    
    print()

if tries_used == max_tries and guess != secret_number:
    print(f"😢 Out of tries! The number was {secret_number}")
```

## 🔤 Game 3: Word Guessing (Like Hangman!)

Can you guess the secret word letter by letter?


```python
# Word Guessing Game!
import random

print("=== WORD GUESSER ===")
print("Guess the word letter by letter!\n")

word_list = ["python", "robot", "dragon", "castle", "wizard", "pirate"]
secret_word = random.choice(word_list)

guessed_letters = []
max_wrong = 6
wrong_guesses = 0

def display_word():
    display = ""
    for letter in secret_word:
        if letter in guessed_letters:
            display += letter + " "
        else:
            display += "_ "
    return display

print(f"The word has {len(secret_word)} letters!")

while wrong_guesses < max_wrong:
    print(f"\n{display_word()}")
    print(f"Wrong guesses: {wrong_guesses}/{max_wrong}")
    print(f"Guessed letters: {', '.join(guessed_letters)}")
    
    print("Guess a letter: ")
    guess = input("Guess a letter: ").lower()
    
    if len(guess) != 1:
        print("Please guess ONE letter!")
        continue
    
    if guess in guessed_letters:
        print("You already guessed that!")
        continue
    
    guessed_letters.append(guess)
    
    if guess in secret_word:
        print("✅ Correct!")
        
        # Check if word is complete
        if all(letter in guessed_letters for letter in secret_word):
            print(f"\n🎉 YOU WIN! The word was: {secret_word}")
            break
    else:
        print("❌ Wrong!")
        wrong_guesses += 1

if wrong_guesses == max_wrong:
    print(f"\n😢 Game Over! The word was: {secret_word}")
```

## 🎯 Game Mechanics Breakdown!

Every game has core **mechanics** - the rules and systems that make it work!

Let's break down common game mechanics:
- **Health System** - tracking player life
- **Inventory** - storing items
- **Score System** - tracking points
- **Turn-Based** - taking turns
- **Random Events** - surprises!


```python
# SIMPLE RPG SYSTEM - Building Blocks!
import random

print("⚔️ RPG MECHANICS DEMO ⚔️\n")

# 1. Character Stats
player = {
    "name": "Hero",
    "health": 100,
    "max_health": 100,
    "attack": 15,
    "defense": 5,
    "level": 1,
    "experience": 0,
    "inventory": ["Potion", "Bread"]
}

# 2. Display Stats Function
def show_stats():
    print(f"\n{'='*40}")
    print(f"Character: {player['name']} (Level {player['level']})")
    print(f"Health: {player['health']}/{player['max_health']}")
    print(f"Attack: {player['attack']} | Defense: {player['defense']}")
    print(f"Experience: {player['experience']}")
    print(f"Inventory: {', '.join(player['inventory'])}")
    print(f"{'='*40}\n")

# 3. Combat System
def combat(enemy_name, enemy_health, enemy_attack):
    print(f"\n⚔️ A wild {enemy_name} appears!")
    
    while enemy_health > 0 and player['health'] > 0:
        print(f"\n{enemy_name} HP: {enemy_health}")
        print(f"Your HP: {player['health']}")
        
        print("(A)ttack or (R)un? ")
        action = input("(A)ttack or (R)un? ").lower()
        
        if action == 'a':
            # Player attacks
            damage = player['attack'] + random.randint(-3, 3)
            enemy_health -= damage
            print(f"You deal {damage} damage!")
            
            if enemy_health <= 0:
                print(f"✅ You defeated the {enemy_name}!")
                xp_gain = 25
                player['experience'] += xp_gain
                print(f"+{xp_gain} XP!")
                return True
            
            # Enemy attacks back
            enemy_damage = enemy_attack - player['defense'] + random.randint(-2, 2)
            if enemy_damage < 0:
                enemy_damage = 0
            player['health'] -= enemy_damage
            print(f"{enemy_name} deals {enemy_damage} damage!")
            
            if player['health'] <= 0:
                print("💀 You were defeated!")
                return False
        elif action == 'r':
            print("🏃 You ran away!")
            return False

# Demo the system
show_stats()
combat("Goblin", 30, 8)
show_stats()
```

## 🌟 Challenge Corner: Dungeon Crawler!

Build a complete dungeon exploration game with:
- Multiple rooms to explore
- Enemies to fight
- Treasures to find
- Health and inventory system
- Victory condition!


```python
# DUNGEON CRAWLER GAME!
import random

print("="*50)
print("🏰 DUNGEON CRAWLER 🏰")
print("="*50)

# Player stats
player = {
    "health": 100,
    "gold": 0,
    "inventory": [],
    "position": "entrance"
}

# Game state
game_over = False
has_key = False

def show_stats():
    print(f"\n💚 Health: {player['health']} | 💰 Gold: {player['gold']}")
    if player['inventory']:
        print(f"📦 Inventory: {', '.join(player['inventory'])}")

# Room 1: Entrance
def entrance():
    global has_key
    print("\n" + "="*50)
    print("📍 THE ENTRANCE")
    print("="*50)
    print("You stand in a dark dungeon entrance.")
    print("1. Go North (Treasure Room)")
    print("2. Go East (Monster Room)")
    print("3. Go West (Armory)")
    
    print("\nYour choice: ")
    choice = input("\nYour choice: ")
    
    if choice == "1":
        treasure_room()
    elif choice == "2":
        monster_room()
    elif choice == "3":
        armory()
    else:
        print("Invalid choice!")
        entrance()

# Room 2: Treasure Room
def treasure_room():
    global has_key, game_over
    print("\n✨ You found the TREASURE ROOM!")
    
    if has_key:
        print("You use the key to open the chest!")
        gold_found = random.randint(50, 100)
        player['gold'] += gold_found
        print(f"💰 You found {gold_found} gold!")
        print("\n🎉 YOU WIN! You escaped with the treasure!")
        game_over = True
    else:
        print("The chest is locked! You need a key.")
        entrance()

# Room 3: Monster Room
def monster_room():
    print("\n👹 A MONSTER blocks your path!")
    enemy_health = 50
    
    while enemy_health > 0 and player['health'] > 0:
        print(f"\nMonster HP: {enemy_health}")
        show_stats()
        
        print("(A)ttack, (R)un, or use (P)otion? ")
        action = input("(A)ttack, (R)un, or use (P)otion? ").lower()
        
        if action == 'a':
            damage = random.randint(10, 20)
            enemy_health -= damage
            print(f"⚔️ You deal {damage} damage!")
            
            if enemy_health > 0:
                enemy_damage = random.randint(5, 15)
                player['health'] -= enemy_damage
                print(f"💥 Monster deals {enemy_damage} damage!")
        
        elif action == 'p':
            if "Potion" in player['inventory']:
                player['health'] += 30
                if player['health'] > 100:
                    player['health'] = 100
                player['inventory'].remove("Potion")
                print("💚 Used potion! +30 health")
            else:
                print("You have no potions!")
        
        elif action == 'r':
            print("🏃 You ran back!")
            entrance()
            return
    
    if enemy_health <= 0:
        print("✅ Monster defeated!")
        player['gold'] += 25
        print("+25 gold!")
        entrance()
    else:
        print("💀 You were defeated!")
        global game_over
        game_over = True

# Room 4: Armory
def armory():
    global has_key
    print("\n⚔️ THE ARMORY")
    print("You find weapons and items!")
    print("1. Take Potion")
    print("2. Take Key")
    print("3. Go Back")
    
    print("\nYour choice: ")
    choice = input("\nYour choice: ")
    
    if choice == "1":
        player['inventory'].append("Potion")
        print("✅ Took a potion!")
    elif choice == "2":
        has_key = True
        player['inventory'].append("Key")
        print("✅ Took the key!")
    
    entrance()

# Start game
print("\nYou enter a mysterious dungeon...")
show_stats()

# Main game loop
while not game_over and player['health'] > 0:
    entrance()
    if game_over:
        break

# Game end
print("\n" + "="*50)
print("GAME OVER")
print("="*50)
show_stats()
print("\nThanks for playing!")
```
