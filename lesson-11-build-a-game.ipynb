## 🔢 Game 2: Advanced Number GuessingRemember the number guessing game? Let's make it MUCH better!

```python
# Advanced Number Guessing Game!import randomprint("=== ULTIMATE NUMBER GUESSER ===")print("I'm thinking of a number between 1 and 50!")print("You have 7 tries to guess it!\n")secret_number = random.randint(1, 50)max_tries = 7tries_used = 0while tries_used < max_tries:    tries_left = max_tries - tries_used    print(f"Tries left: {tries_left}")        guess = int(input("Your guess: "))    tries_used += 1        if guess == secret_number:        print(f"\n🎉 AMAZING! You got it in {tries_used} tries!")        break    elif guess < secret_number:        difference = secret_number - guess        if difference <= 5:            print("🔥 Very close! Go HIGHER!")        elif difference <= 10:            print("📈 Close! Go higher!")        else:            print("📈 Too low! Go much higher!")    elif guess > secret_number:        difference = guess - secret_number        if difference <= 5:            print("🔥 Very close! Go LOWER!")        elif difference <= 10:            print("📉 Close! Go lower!")        else:            print("📉 Too high! Go much lower!")        print()if tries_used == max_tries and guess != secret_number:    print(f"😢 Out of tries! The number was {secret_number}")
```

## 🔤 Game 3: Word Guessing (Like Hangman!)Can you guess the secret word letter by letter?

```python
# Word Guessing Game!import randomprint("=== WORD GUESSER ===")print("Guess the word letter by letter!\n")word_list = ["python", "robot", "dragon", "castle", "wizard", "pirate"]secret_word = random.choice(word_list)guessed_letters = []max_wrong = 6wrong_guesses = 0def display_word():    display = ""    for letter in secret_word:        if letter in guessed_letters:            display += letter + " "        else:            display += "_ "    return displayprint(f"The word has {len(secret_word)} letters!")while wrong_guesses < max_wrong:    print(f"\n{display_word()}")    print(f"Wrong guesses: {wrong_guesses}/{max_wrong}")    print(f"Guessed letters: {', '.join(guessed_letters)}")        guess = input("Guess a letter: ").lower()        if len(guess) != 1:        print("Please guess ONE letter!")        continue        if guess in guessed_letters:        print("You already guessed that!")        continue        guessed_letters.append(guess)        if guess in secret_word:        print("✅ Correct!")                # Check if word is complete        if all(letter in guessed_letters for letter in secret_word):            print(f"\n🎉 YOU WIN! The word was: {secret_word}")            break    else:        print("❌ Wrong!")        wrong_guesses += 1if wrong_guesses == max_wrong:    print(f"\n😢 Game Over! The word was: {secret_word}")
```
