# 📦 Lesson 3: Memory Box (Variables)Hello again! Today we're going to learn about **variables** - they're like special boxes where Python remembers things!Imagine you have a toy box with a label on it. You can put toys inside and the label tells you what's in the box. Variables work the same way!

## 🏷️ Your First VariableA variable has two parts:1. A **name** (the label on the box)2. A **value** (what's inside the box)We create a variable using the `=` sign, like this:`name = "Owen"`This means: "Create a box called 'name' and put 'Owen' inside it"

```python
# Create your own story!hero = "Robot"villain = "Dragon"treasure = "golden pizza"place = "castle"print("Once upon a time...")print(hero, "went to the", place)print("There was a", villain, "guarding the", treasure)print(hero, "defeated the", villain)print("And got the", treasure, "!")print("The end!")# Try changing the variables above to make a new story!
```

## 🔄 Variables Can Change!The cool thing about variables is you can change what's inside them - just like taking toys out of a box and putting new ones in!

```python
# Watch the variable change!pet = "dog"print("I have a", pet)pet = "cat"print("Now I have a", pet)pet = "dragon"print("Actually, I have a", pet, "!")
```

## 🔢 Variables with NumbersVariables are super helpful for math! You can do calculations with them:

```python
# Math with variables!apples = 5oranges = 3print("I have", apples, "apples")print("I have", oranges, "oranges")total_fruit = apples + orangesprint("Total fruit:", total_fruit)# Let's get more apples!apples = apples + 10print("\nNow I have", apples, "apples!")
```
