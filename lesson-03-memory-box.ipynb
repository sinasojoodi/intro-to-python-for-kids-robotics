## 📛 Variable Naming Rules!

There are important rules for naming variables in Python! Think of it like naming your pet - some names work, some don't!

### ✅ GOOD Names (These Work!):
- Use letters, numbers, and underscores `_`
- Start with a letter or underscore (not a number!)
- Use lowercase for most variables
- Use descriptive names: `player_score` is better than `ps`

### ❌ BAD Names (These Don't Work!):
- Can't use spaces: `my variable` ❌ use `my_variable` ✅
- Can't start with numbers: `2cool` ❌ use `cool2` ✅
- Can't use special characters: `my-name` or `total$` ❌
- Can't use Python keywords: `print`, `for`, `if` ❌

Let's see examples!


```python
# GOOD variable names ✅
player_name = "Alex"
player_score = 100
total_points = 50
my_age = 7
favorite_number = 42

print("Good names work great!")
print("Player:", player_name, "Score:", player_score)

# These would cause errors (don't uncomment them!):
# 2fast = 50  # Can't start with number
# my-name = "Sam"  # Can't use dash
# total score = 100  # Can't have space
```

## 🎯 Multiple Assignment: Super Shortcut!

Did you know you can create multiple variables at once? It's like filling several boxes in one line!

You can do: `a, b, c = 1, 2, 3`

This creates three variables all at once! Pretty cool!


```python
# Create three variables at once!
x, y, z = 10, 20, 30
print("x =", x)
print("y =", y)
print("z =", z)

print("\n---")

# This works with strings too!
first_name, last_name, age = "Owen", "Smith", 7
print("Name:", first_name, last_name)
print("Age:", age)

print("\n---")

# You can even assign the same value to multiple variables!
a = b = c = 100
print("a =", a, "b =", b, "c =", c)
```

## ⚡ Compound Operators: Quick Math Shortcuts!

There's a faster way to change variables! Instead of writing `score = score + 10`, you can use **compound operators**!

- `+=` means "add to this variable"
- `-=` means "subtract from this variable"
- `*=` means "multiply this variable"
- `/=` means "divide this variable"

These save time and make code easier to read!


```python
# Using += (add and assign)
score = 100
print("Starting score:", score)

score += 10  # Same as: score = score + 10
print("After gaining 10 points:", score)

score += 25  # Gain more points!
print("After gaining 25 more:", score)

print("\n---")

# Using -= (subtract and assign)
health = 100
print("Starting health:", health)

health -= 20  # Same as: health = health - 20
print("After taking damage:", health)

print("\n---")

# Using *= (multiply and assign)
power = 5
print("Starting power:", power)

power *= 3  # Same as: power = power * 3
print("After power-up (×3):", power)

print("\n---")

# Using /= (divide and assign)
money = 100
print("You have $", money)

money /= 2  # Same as: money = money / 2
print("After sharing with friend:", money)
```

## 🔄 Variable Swapping: The Magic Trick!

What if you have two boxes and want to switch what's inside them?

If you have `a = 5` and `b = 10`, how do you make them switch? You need a temporary box!

This is called **variable swapping** and it's like a magic trick!


```python
# Method 1: Using a temporary variable (like a third box!)
print("🎩 Variable Swapping Magic! 🎩\n")

a = 5
b = 10
print("Before swap:")
print("a =", a)
print("b =", b)

# The swap magic!
temp = a      # Put a's value in temp box
a = b         # Put b's value in a
b = temp      # Put temp's value in b

print("\nAfter swap:")
print("a =", a)
print("b =", b)

print("\n" + "="*30 + "\n")

# Method 2: Python's special shortcut!
x = "cat"
y = "dog"
print("Before swap:")
print("x =", x)
print("y =", y)

x, y = y, x  # Python can do it in one line!

print("\nAfter swap:")
print("x =", x)
print("y =", y)
```
