## 🎯 Order of Operations: PEMDAS!

When you have a math problem with many operations, Python follows special rules called **Order of Operations**!

Remember **PEMDAS**:
- **P**arentheses first `( )`
- **E**xponents (powers) `**`
- **M**ultiplication and **D**ivision `*` `/` (left to right)
- **A**ddition and **S**ubtraction `+` `-` (left to right)

Let's see it in action!


```python
# Without parentheses - multiplication happens first!
print("5 + 3 * 2 =")
print(5 + 3 * 2)  # 3 * 2 = 6, then 5 + 6 = 11

print("\n---")

# With parentheses - what's inside happens first!
print("(5 + 3) * 2 =")
print((5 + 3) * 2)  # 5 + 3 = 8, then 8 * 2 = 16

print("\n---")

# More complex example
print("2 + 3 * 4 - 5 =")
print(2 + 3 * 4 - 5)  # 3 * 4 = 12, then 2 + 12 - 5 = 9
```
