## ⚡ Super Math Powers!Python has special math powers! Check these out:- `**` means "to the power of" (5 ** 2 means 5 × 5)- `%` means "remainder" (10 % 3 means what's left after dividing)These are like secret math tricks!

```python
# Secret math tricks!print("What is 5 squared (5 × 5)?")print(5 ** 2)print("\nWhat is 2 to the power of 4 (2×2×2×2)?")print(2 ** 4)print("\nWhat's left when we divide 10 by 3?")print(10 % 3)
```
