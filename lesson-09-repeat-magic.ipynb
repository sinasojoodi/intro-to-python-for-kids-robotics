## 🔢 The for Loop with range()

The `for` loop is perfect when you know HOW MANY times you want to repeat something.

**Format:**
```python
for i in range(number):
    do this repeatedly
```

`range(5)` means: 0, 1, 2, 3, 4 (five numbers!)


````python
# Break and continue!

print("Finding a number:")
for i in range(1, 11):
    if i == 7:
        print(f"Found {i}! Stopping!")
        break
    print(i)

print("\n--- Skip the odd numbers ---")
for i in range(1, 11):
    if i % 2 == 1:  # If odd
        continue  # Skip it!
    print(i)  # Only even numbers print!

print("\n--- Secret code ---")
attempts = 0
while attempts < 5:
    code = input("Enter code (hint: it's 1234): ")
    attempts += 1
    
    if code == "1234":
        print("✅ Correct! Access granted!")
        break
    else:
        print(f"❌ Wrong! {5 - attempts} tries left")
````

## 🪆 Nested Loops

You can put a loop INSIDE another loop! This is called a **nested loop**.

It's like:
- For each day of the week
  - For each hour of the day
    - Do something


````python
# Nested loops!

print("Multiplication table:")
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i} x {j} = {i*j}")
    print("---")  # Separator

print("\n--- Pattern maker ---")
for row in range(5):
    for col in range(row + 1):
        print("⭐", end="")
    print()  # New line

print("\n--- Number pyramid ---")
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

## 🎨 ASCII Art with Loops!

Let's make cool patterns using loops!


````python
# ASCII art patterns!

print("Pattern 1:")
for i in range(5):
    print("*" * (i + 1))

print("\nPattern 2:")
for i in range(5):
    spaces = " " * (5 - i - 1)
    stars = "*" * (2 * i + 1)
    print(spaces + stars)

print("\nPattern 3 - Christmas Tree:")
for i in range(5):
    spaces = " " * (5 - i)
    stars = "*" * (2 * i + 1)
    print(spaces + stars)
print("  " + "  |||")  # Trunk

print("\nPattern 4 - Box:")
size = 7
for i in range(size):
    if i == 0 or i == size - 1:
        print("*" * size)
    else:
        print("*" + " " * (size - 2) + "*")
````

## 🎯 range() Tricks

`range()` is super flexible!

- `range(5)` → 0, 1, 2, 3, 4
- `range(1, 6)` → 1, 2, 3, 4, 5
- `range(0, 10, 2)` → 0, 2, 4, 6, 8 (skip by 2!)
- `range(10, 0, -1)` → 10, 9, 8, ... 1 (backwards!)


````python
# Exploring range!

print("From 1 to 10:")
for num in range(1, 11):
    print(num, end=" ")  # Print on same line!

print("\n\nEven numbers:")
for num in range(0, 21, 2):
    print(num, end=" ")

print("\n\nCount by 5s:")
for num in range(0, 51, 5):
    print(num, end=" ")

print("\n\nTimes table for 3:")
for i in range(1, 11):
    print(f"3 x {i} = {3 * i}")
````

## 📝 Looping Through Lists

You can loop through items in a list! Python will look at each item one by one:


````python
# Loop through lists!

animals = ["dog", "cat", "rabbit", "hamster", "fish"]

print("My favorite animals:")
for animal in animals:
    print(f"I like {animal}s!")

print("\n--- Favorite foods ---")
foods = ["pizza", "ice cream", "tacos", "cookies"]

for food in foods:
    print(f"Yum! {food} is delicious!")

print("\n--- Letter by letter ---")
name = "PYTHON"

for letter in name:
    print(letter)
````

## ♾️ The while Loop

The `while` loop keeps repeating AS LONG AS something is True!

**Format:**
```python
while condition:
    do this
```

Be careful! If the condition never becomes False, it loops FOREVER!


````python
# While loop examples!

count = 1

print("Counting with while:")
while count <= 5:
    print(count)
    count = count + 1  # Important! Update the counter!

print("\n--- Eating cookies ---")
cookies = 10

while cookies > 0:
    print(f"{cookies} cookies left!")
    cookies = cookies - 1
    if cookies > 0:
        print("Nom nom! *eating a cookie*")

print("All cookies gone! 🍪")
````

## 🔢 enumerate(): Loop with Position Numbers!

Sometimes you want to know BOTH the item AND its position in a list. That's what `enumerate()` does!

It gives you two things:
- The **index** (position number): 0, 1, 2, 3...
- The **item** itself

Super useful!


````python
# Using enumerate!
fruits = ["apple", "banana", "orange", "grape", "mango"]

print("🍎 FRUIT LIST 🍎")
for index, fruit in enumerate(fruits):
    print(f"{index + 1}. {fruit.title()}")

print("\n---")

# Useful for numbered lists!
tasks = ["Do homework", "Clean room", "Practice coding", "Feed pet"]
print("📝 TODO LIST 📝")
for num, task in enumerate(tasks, start=1):  # Start counting from 1!
    print(f"Task {num}: {task}")

print("\n---")

# Finding position of specific items
colors = ["red", "blue", "green", "yellow", "purple"]
print("Finding 'green':")
for index, color in enumerate(colors):
    if color == "green":
        print(f"Found {color} at position {index}!")
````

## 📊 Accumulator Pattern: Building Up Values!

An **accumulator** is a variable that "accumulates" (collects/adds up) values as you loop!

Common uses:
- Adding up numbers (total)
- Counting things (counter)
- Building strings (concatenation)
- Creating lists (append)


````python
# Accumulator examples!

# Example 1: Sum accumulator
print("Adding up numbers 1 to 10:")
total = 0  # Start at zero!
for num in range(1, 11):
    total += num
    print(f"Adding {num}: total is now {total}")
print(f"Final total: {total}")

print("\n---")

# Example 2: Counter accumulator
print("\nCounting vowels:")
word = "programming"
vowel_count = 0
for letter in word:
    if letter in "aeiou":
        vowel_count += 1
        print(f"Found vowel: {letter}")
print(f"Total vowels: {vowel_count}")

print("\n---")

# Example 3: String accumulator
print("\nBuilding a message:")
message = ""
words = ["Python", "is", "super", "fun!"]
for word in words:
    message += word + " "
print(f"Message: {message}")

print("\n---")

# Example 4: List accumulator
print("\nCollecting even numbers:")
evens = []
for num in range(1, 21):
    if num % 2 == 0:
        evens.append(num)
print(f"Even numbers: {evens}")
````

## 🐛 Debug This!

These loop programs have bugs! Can you find and fix them?

**Hint:** Look for range errors, wrong indentation, infinite loops, and accumulator mistakes!


````python
# Bug 1: Wrong range
# Should print 1 to 5, but starts at 0!
for i in range(5):  # Should be range(1, 6)
    print(i)

# Bug 2: Infinite loop
# This will never stop!
count = 1
while count <= 5:
    print(count)
    # Missing: count += 1

# Bug 3: Wrong indentation
for i in range(3):
    print("Starting loop")
print("Loop iteration", i)  # This should be indented!

# Bug 4: Accumulator not initialized
# total is never set to 0!
for num in range(1, 6):
    total += num  # Error! total doesn't exist yet!
print(total)

# Bug 5: Off-by-one error
# Trying to print 1 to 10, but missing 10!
for num in range(1, 10):  # Should be range(1, 11)
    print(num)

# Bug 6: Wrong counter update
score = 0
for i in range(5):
    score = score + i  # Should be += not reassignment
print("Score:", i)  # Wrong! Should print score, not i!
````

## 🌟 Challenge Corner: Multiplication Quiz Game!

Ready for the ultimate challenge? Build a multiplication quiz that:
- Uses random numbers
- Loops through 10 questions
- Keeps track of correct answers
- Shows final score and percentage
- Gives encouraging messages!


````python
# MULTIPLICATION QUIZ GAME!
import random

print("=" * 50)
print("🧮 MULTIPLICATION QUIZ MASTER 🧮")
print("=" * 50)
print("Test your multiplication skills!")
print("You'll get 10 questions.\n")

correct_count = 0
total_questions = 10

for question_num in range(1, total_questions + 1):
    # Generate random numbers
    num1 = random.randint(1, 12)
    num2 = random.randint(1, 12)
    correct_answer = num1 * num2
    
    # Ask the question
    print(f"\nQuestion {question_num}:")
    user_answer = int(input(f"What is {num1} × {num2}? "))
    
    # Check the answer
    if user_answer == correct_answer:
        print("✅ Correct! Great job!")
        correct_count += 1
    else:
        print(f"❌ Not quite. The answer is {correct_answer}")

# Calculate percentage
percentage = (correct_count / total_questions) * 100

# Show results
print("\n" + "=" * 50)
print("FINAL RESULTS:")
print("=" * 50)
print(f"You got {correct_count} out of {total_questions} correct!")
print(f"That's {percentage}%!")

# Encouraging message based on score
if percentage == 100:
    print("\n🏆 PERFECT SCORE! You're a multiplication master!")
elif percentage >= 80:
    print("\n🌟 Excellent work! You really know your times tables!")
elif percentage >= 60:
    print("\n👍 Good job! Keep practicing!")
else:
    print("\n💪 Keep trying! Practice makes perfect!")

print("=" * 50)
````
