## 🔢 The for Loop with range()The `for` loop is perfect when you know HOW MANY times you want to repeat something.**Format:**```pythonfor i in range(number):    do this repeatedly````range(5)` means: 0, 1, 2, 3, 4 (five numbers!)

`````python
# Break and continue!print("Finding a number:")for i in range(1, 11):    if i == 7:        print(f"Found {i}! Stopping!")        break    print(i)print("\n--- Skip the odd numbers ---")for i in range(1, 11):    if i % 2 == 1:  # If odd        continue  # Skip it!    print(i)  # Only even numbers print!print("\n--- Secret code ---")attempts = 0while attempts < 5:    code = input("Enter code (hint: it's 1234): ")    attempts += 1        if code == "1234":        print("✅ Correct! Access granted!")        break    else:        print(f"❌ Wrong! {5 - attempts} tries left")
`````

## 🪆 Nested LoopsYou can put a loop INSIDE another loop! This is called a **nested loop**.It's like:- For each day of the week  - For each hour of the day    - Do something

`````python
# Nested loops!print("Multiplication table:")for i in range(1, 6):    for j in range(1, 6):        print(f"{i} x {j} = {i*j}")    print("---")  # Separatorprint("\n--- Pattern maker ---")for row in range(5):    for col in range(row + 1):        print("⭐", end="")    print()  # New lineprint("\n--- Number pyramid ---")for i in range(1, 6):    for j in range(1, i + 1):        print(j, end=" ")    print()
`````

## 🎯 range() Tricks`range()` is super flexible!- `range(5)` → 0, 1, 2, 3, 4- `range(1, 6)` → 1, 2, 3, 4, 5- `range(0, 10, 2)` → 0, 2, 4, 6, 8 (skip by 2!)- `range(10, 0, -1)` → 10, 9, 8, ... 1 (backwards!)

`````python
# Exploring range!print("From 1 to 10:")for num in range(1, 11):    print(num, end=" ")  # Print on same line!print("\n\nEven numbers:")for num in range(0, 21, 2):    print(num, end=" ")print("\n\nCount by 5s:")for num in range(0, 51, 5):    print(num, end=" ")print("\n\nTimes table for 3:")for i in range(1, 11):    print(f"3 x {i} = {3 * i}")
`````

## 📝 Looping Through ListsYou can loop through items in a list! Python will look at each item one by one:

`````python
# Loop through lists!animals = ["dog", "cat", "rabbit", "hamster", "fish"]print("My favorite animals:")for animal in animals:    print(f"I like {animal}s!")print("\n--- Favorite foods ---")foods = ["pizza", "ice cream", "tacos", "cookies"]for food in foods:    print(f"Yum! {food} is delicious!")print("\n--- Letter by letter ---")name = "PYTHON"for letter in name:    print(letter)
`````

## ♾️ The while LoopThe `while` loop keeps repeating AS LONG AS something is True!**Format:**```pythonwhile condition:    do this```Be careful! If the condition never becomes False, it loops FOREVER!

`````python
# While loop examples!count = 1print("Counting with while:")while count <= 5:    print(count)    count = count + 1  # Important! Update the counter!print("\n--- Eating cookies ---")cookies = 10while cookies > 0:    print(f"{cookies} cookies left!")    cookies = cookies - 1    if cookies > 0:        print("Nom nom! *eating a cookie*")print("All cookies gone! 🍪")
`````

## 🛑 Break and ContinueSpecial commands to control loops:- `break` → EXIT the loop immediately- `continue` → SKIP to the next iterationThese are like emergency controls!
